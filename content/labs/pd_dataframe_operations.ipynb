# DataFrame Operations

Master essential Pandas DataFrame operations for data manipulation.

```python
import pandas as pd
import numpy as np
```

```python
# Create DataFrame
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'age': [25, 30, 35, 28],
    'salary': [50000, 60000, 70000, 55000],
    'department': ['Sales', 'IT', 'IT', 'Sales']
})
print(df)
```

```python
# Selection
print(f"Column: {df['name'].tolist()}")
print(f"\nRow by index: {df.iloc[0].to_dict()}")
print(f"\nMultiple columns:\n{df[['name', 'salary']]}")
```

```python
# Filtering
seniors = df[df['age'] > 28]
print(f"Age > 28:\n{seniors}")

it_high_salary = df[(df['department'] == 'IT') & (df['salary'] > 65000)]
print(f"\nIT with salary > 65000:\n{it_high_salary}")
```

```python
# Adding columns
df['bonus'] = df['salary'] * 0.1
df['senior'] = df['age'] > 30
print(df)
```

```python
# Apply functions
df['name_upper'] = df['name'].apply(str.upper)
df['tax'] = df['salary'].apply(lambda x: x * 0.2 if x > 55000 else x * 0.15)
print(df[['name', 'salary', 'tax']])
```

```python
# Sorting
print(df.sort_values('salary', ascending=False)[['name', 'salary']])
```

## Key Observations
- df['col'] returns Series, df[['col']] returns DataFrame
- iloc for position, loc for labels
- Chain conditions with & (and) | (or), use parentheses
- apply() for row/column-wise operations

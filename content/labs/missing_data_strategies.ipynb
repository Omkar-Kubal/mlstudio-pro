# Missing Data Strategies

Compare deletion, imputation, and indicator methods for missing data.

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
np.random.seed(42)

X = np.random.randn(100, 3)
X[np.random.random(X.shape) < 0.2] = np.nan
df = pd.DataFrame(X, columns=['A', 'B', 'C'])
print(f"Missing: {df.isnull().sum().to_dict()}")
```

```python
# Mean imputation
mean_imp = SimpleImputer(strategy='mean')
df_mean = pd.DataFrame(mean_imp.fit_transform(df), columns=df.columns)
print(f"After mean imputation: {df_mean.isnull().sum().sum()} missing")

# Median imputation
median_imp = SimpleImputer(strategy='median')
df_median = pd.DataFrame(median_imp.fit_transform(df), columns=df.columns)

# Compare
print(f"Original std: {df.std().round(2).tolist()}")
print(f"Mean imputed std: {df_mean.std().round(2).tolist()}")
```

## Key Observations
- Deletion: loses data, only if MCAR
- Mean/median: reduces variance
- Model-based (KNN, iterative): preserves relationships better

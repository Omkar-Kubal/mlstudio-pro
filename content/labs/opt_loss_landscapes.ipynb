# Loss Landscapes

Visualize loss functions as surfaces in parameter space to understand optimization challenges.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
# Create parameter grid
w1 = np.linspace(-3, 3, 100)
w2 = np.linspace(-3, 3, 100)
W1, W2 = np.meshgrid(w1, w2)
```

```python
# Simple convex loss (quadratic bowl)
# L(w1, w2) = w1² + w2² (minimum at origin)
L_convex = W1**2 + W2**2

# Elongated loss (different curvature per dimension)
L_elongated = W1**2 + 10 * W2**2

# Non-convex loss (multiple minima)
L_nonconvex = np.sin(W1) * np.cos(W2) + 0.1*(W1**2 + W2**2)
```

```python
fig = plt.figure(figsize=(15, 5))

# 3D surface plots
for i, (L, title) in enumerate([
    (L_convex, 'Convex (w1² + w2²)'),
    (L_elongated, 'Elongated (w1² + 10w2²)'),
    (L_nonconvex, 'Non-convex')
]):
    ax = fig.add_subplot(1, 3, i+1, projection='3d')
    ax.plot_surface(W1, W2, L, cmap='viridis', alpha=0.8)
    ax.set_xlabel('w1')
    ax.set_ylabel('w2')
    ax.set_zlabel('Loss')
    ax.set_title(title)

plt.tight_layout()
plt.show()
```

```python
# Contour plots (2D view of loss landscape)
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

for ax, (L, title) in zip(axes, [
    (L_convex, 'Convex'),
    (L_elongated, 'Elongated'),
    (L_nonconvex, 'Non-convex')
]):
    cs = ax.contour(W1, W2, L, levels=20, cmap='viridis')
    ax.clabel(cs, inline=True, fontsize=8)
    ax.set_xlabel('w1')
    ax.set_ylabel('w2')
    ax.set_title(title)
    ax.set_aspect('equal')

plt.tight_layout()
plt.show()
```

```python
# Rosenbrock function (classic optimization test)
# L = (1-w1)² + 100(w2 - w1²)²
# Global minimum at (1, 1), but hard to find
w1_r = np.linspace(-2, 2, 200)
w2_r = np.linspace(-1, 3, 200)
W1_r, W2_r = np.meshgrid(w1_r, w2_r)

L_rosenbrock = (1 - W1_r)**2 + 100 * (W2_r - W1_r**2)**2

plt.figure(figsize=(8, 6))
plt.contour(W1_r, W2_r, np.log(L_rosenbrock + 1), levels=30, cmap='viridis')
plt.scatter([1], [1], color='red', s=100, marker='*', label='Global min (1,1)')
plt.xlabel('w1')
plt.ylabel('w2')
plt.title('Rosenbrock Function (log scale)')
plt.legend()
plt.show()
```

```python
# Linear regression loss landscape
np.random.seed(42)
X = np.random.randn(50, 1)
y = 2 * X.ravel() + 1 + np.random.randn(50) * 0.5  # True: slope=2, intercept=1

def mse_loss(w0, w1, X, y):
    predictions = w0 + w1 * X.ravel()
    return np.mean((y - predictions) ** 2)

w0_range = np.linspace(-1, 3, 50)
w1_range = np.linspace(0, 4, 50)
W0, W1_grid = np.meshgrid(w0_range, w1_range)

L_mse = np.zeros_like(W0)
for i in range(len(w0_range)):
    for j in range(len(w1_range)):
        L_mse[j, i] = mse_loss(w0_range[i], w1_range[j], X, y)

plt.figure(figsize=(8, 6))
cs = plt.contour(W0, W1_grid, L_mse, levels=20, cmap='viridis')
plt.scatter([1], [2], color='red', s=100, marker='*', label='True params')
plt.xlabel('w0 (intercept)')
plt.ylabel('w1 (slope)')
plt.title('MSE Loss Landscape for Linear Regression')
plt.colorbar(cs, label='MSE')
plt.legend()
plt.show()
```

## Key Observations

- Convex loss = single global minimum, optimization is straightforward.
- Elongated contours = different learning rates needed per dimension.
- Non-convex loss = multiple local minima, optimization can get stuck.
- Linear regression has convex (quadratic bowl) loss landscape.
- Contour plots reveal optimization difficulty (narrow valleys, saddles).

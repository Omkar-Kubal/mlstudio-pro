# Dot Product, Norm, and Similarity

Explore the dot product as a measure of alignment, compute norms as vector magnitude, and use cosine similarity.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Two vectors
v = np.array([3, 4])
w = np.array([4, 3])

# Dot product
dot_product = np.dot(v, w)
print(f"v = {v}, w = {w}")
print(f"v · w = {dot_product}")
print(f"Manual: 3*4 + 4*3 = {3*4 + 4*3}")
```

```python
# Norm (magnitude)
norm_v = np.linalg.norm(v)
norm_w = np.linalg.norm(w)

print(f"\n||v|| = {norm_v:.4f}")
print(f"||w|| = {norm_w:.4f}")
print(f"Manual ||v||: sqrt(3² + 4²) = {np.sqrt(3**2 + 4**2)}")
```

```python
# Cosine similarity
cos_sim = dot_product / (norm_v * norm_w)
angle_rad = np.arccos(cos_sim)
angle_deg = np.degrees(angle_rad)

print(f"\nCosine similarity: {cos_sim:.4f}")
print(f"Angle between vectors: {angle_deg:.2f}°")
```

```python
# Special cases
identical = np.array([1, 0])
orthogonal = np.array([0, 1])
opposite = np.array([-1, 0])

print("\nSpecial angle cases:")
for name, u in [("identical", identical), ("orthogonal", orthogonal), ("opposite", opposite)]:
    dot = np.dot(identical, u)
    cos = dot / (np.linalg.norm(identical) * np.linalg.norm(u))
    print(f"  [1,0] · {u}: dot={dot}, cos={cos:.2f}")
```

```python
# Visualize angles
fig, ax = plt.subplots(figsize=(8, 8))

origin = np.array([0, 0])
reference = np.array([1, 0])

angles = [0, 30, 60, 90, 120, 180]
colors = plt.cm.viridis(np.linspace(0, 1, len(angles)))

for ang, color in zip(angles, colors):
    rad = np.radians(ang)
    vec = np.array([np.cos(rad), np.sin(rad)])
    ax.quiver(*origin, *vec, angles='xy', scale_units='xy', scale=1, color=color, 
              label=f'{ang}° (cos={np.cos(rad):.2f})')

ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-0.5, 1.5)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.axhline(0, color='black', linewidth=0.5)
ax.axvline(0, color='black', linewidth=0.5)
ax.legend(loc='upper right')
ax.set_title('Cosine Similarity at Different Angles')
plt.show()
```

```python
# Document similarity example
# Word counts: ["machine", "learning", "data", "science"]
doc1 = np.array([3, 5, 2, 1])  # ML-focused
doc2 = np.array([4, 4, 3, 2])  # Similar topic
doc3 = np.array([0, 0, 8, 1])  # Data-focused, less ML

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print("\nDocument similarity (word count vectors):")
print(f"  doc1 vs doc2: {cosine_similarity(doc1, doc2):.4f}")
print(f"  doc1 vs doc3: {cosine_similarity(doc1, doc3):.4f}")
print(f"  doc2 vs doc3: {cosine_similarity(doc2, doc3):.4f}")
```

```python
# L1 vs L2 norms
x = np.array([3, 4])
l1_norm = np.linalg.norm(x, ord=1)  # |3| + |4| = 7
l2_norm = np.linalg.norm(x, ord=2)  # sqrt(9 + 16) = 5
linf_norm = np.linalg.norm(x, ord=np.inf)  # max(|3|, |4|) = 4

print(f"\nDifferent norms for x = {x}:")
print(f"  L1 (Manhattan): {l1_norm}")
print(f"  L2 (Euclidean): {l2_norm}")
print(f"  L∞ (Max):       {linf_norm}")
```

## Key Observations

- Dot product measures how aligned two vectors are.
- Norm gives the "length" or magnitude of a vector.
- Cosine similarity = cos(θ) ∈ [-1, 1], independent of magnitude.
- Orthogonal vectors have dot product = 0.
- Used in recommendation systems, NLP, and similarity search.

# Optimizer Comparison

Visualize SGD, Momentum, and Adam on a loss surface.

```python
import numpy as np
import matplotlib.pyplot as plt

def loss(w): return w[0]**2 + 10*w[1]**2  # Elongated bowl
def grad(w): return np.array([2*w[0], 20*w[1]])
```

```python
def sgd(w, lr=0.1, steps=50):
    path = [w.copy()]
    for _ in range(steps):
        w = w - lr * grad(w)
        path.append(w.copy())
    return np.array(path)

def momentum(w, lr=0.1, beta=0.9, steps=50):
    path = [w.copy()]
    v = np.zeros_like(w)
    for _ in range(steps):
        v = beta * v + lr * grad(w)
        w = w - v
        path.append(w.copy())
    return np.array(path)

def adam(w, lr=0.1, beta1=0.9, beta2=0.999, eps=1e-8, steps=50):
    path = [w.copy()]
    m, v = np.zeros_like(w), np.zeros_like(w)
    for t in range(1, steps+1):
        g = grad(w)
        m = beta1*m + (1-beta1)*g
        v = beta2*v + (1-beta2)*g**2
        m_hat, v_hat = m/(1-beta1**t), v/(1-beta2**t)
        w = w - lr * m_hat / (np.sqrt(v_hat) + eps)
        path.append(w.copy())
    return np.array(path)
```

```python
w0 = np.array([3.0, 3.0])
paths = {'SGD': sgd(w0), 'Momentum': momentum(w0), 'Adam': adam(w0)}

x = np.linspace(-4, 4, 100)
X, Y = np.meshgrid(x, x)
Z = X**2 + 10*Y**2

plt.figure(figsize=(10, 8))
plt.contour(X, Y, Z, levels=30, cmap='viridis', alpha=0.5)
for name, path in paths.items():
    plt.plot(path[:, 0], path[:, 1], '-o', markersize=3, label=name)
plt.scatter(*w0, color='green', s=100, zorder=5, label='Start')
plt.legend()
plt.title('Optimizer Paths')
plt.show()
```

## Key Observations
- SGD: oscillates in high-curvature directions
- Momentum: accumulates velocity, smoother path
- Adam: adaptive learning rates per parameter
- Adam is default for most deep learning

# Image Preprocessing

Common preprocessing: normalization, resizing, augmentation.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
image = np.random.randint(0, 256, (100, 100, 3), dtype=np.uint8)
```

```python
# Normalization
normalized_01 = image / 255.0
normalized_centered = (image / 255.0 - 0.5) * 2  # [-1, 1]

print(f"Original range: [{image.min()}, {image.max()}]")
print(f"Normalized [0,1]: [{normalized_01.min():.2f}, {normalized_01.max():.2f}]")
print(f"Normalized [-1,1]: [{normalized_centered.min():.2f}, {normalized_centered.max():.2f}]")
```

```python
# Simple augmentations
flipped_h = np.fliplr(image)
flipped_v = np.flipud(image)
rotated = np.rot90(image)

fig, axes = plt.subplots(1, 4, figsize=(14, 3))
for ax, img, title in zip(axes, [image, flipped_h, flipped_v, rotated], 
                           ['Original', 'Flip H', 'Flip V', 'Rotate 90']):
    ax.imshow(img)
    ax.set_title(title)
    ax.axis('off')
plt.show()
```

## Key Observations
- Normalize to consistent range for training
- Data augmentation increases effective dataset size
- Common: flips, rotations, crops, color jitter
- Augmentation helps prevent overfitting

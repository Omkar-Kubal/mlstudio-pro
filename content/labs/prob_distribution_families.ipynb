# Distribution Families

Sample and visualize common probability distributions to understand their shapes and use cases.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
np.random.seed(42)
```

```python
n = 10000

# Discrete distributions
bernoulli = np.random.binomial(1, 0.3, n)     # Single trial, p=0.3
binomial = np.random.binomial(20, 0.5, n)    # 20 trials, p=0.5
poisson = np.random.poisson(lam=5, size=n)   # λ=5 events per interval

# Continuous distributions
uniform = np.random.uniform(0, 10, n)
normal = np.random.normal(50, 10, n)
exponential = np.random.exponential(scale=5, size=n)
```

```python
print("DISCRETE DISTRIBUTIONS:")
print(f"  Bernoulli(p=0.3): Mean={np.mean(bernoulli):.3f} (expected 0.3)")
print(f"  Binomial(n=20, p=0.5): Mean={np.mean(binomial):.2f} (expected 10)")
print(f"  Poisson(λ=5): Mean={np.mean(poisson):.2f} (expected 5)")

print("\nCONTINUOUS DISTRIBUTIONS:")
print(f"  Uniform(0, 10): Mean={np.mean(uniform):.2f} (expected 5)")
print(f"  Normal(50, 10): Mean={np.mean(normal):.2f}, Std={np.std(normal):.2f}")
print(f"  Exponential(λ=1/5): Mean={np.mean(exponential):.2f} (expected 5)")
```

```python
fig, axes = plt.subplots(2, 3, figsize=(14, 8))

# Discrete
axes[0, 0].bar([0, 1], [np.mean(bernoulli == 0), np.mean(bernoulli == 1)])
axes[0, 0].set_title('Bernoulli(p=0.3)')
axes[0, 0].set_xticks([0, 1])

vals, counts = np.unique(binomial, return_counts=True)
axes[0, 1].bar(vals, counts / n)
axes[0, 1].set_title('Binomial(n=20, p=0.5)')

vals, counts = np.unique(poisson, return_counts=True)
axes[0, 2].bar(vals, counts / n)
axes[0, 2].set_title('Poisson(λ=5)')

# Continuous
axes[1, 0].hist(uniform, bins=30, density=True, alpha=0.7, edgecolor='black')
axes[1, 0].set_title('Uniform(0, 10)')

axes[1, 1].hist(normal, bins=40, density=True, alpha=0.7, edgecolor='black')
# Overlay theoretical PDF
x = np.linspace(10, 90, 100)
axes[1, 1].plot(x, stats.norm.pdf(x, 50, 10), 'r-', lw=2)
axes[1, 1].set_title('Normal(50, 10)')

axes[1, 2].hist(exponential, bins=40, density=True, alpha=0.7, edgecolor='black')
x = np.linspace(0, 30, 100)
axes[1, 2].plot(x, stats.expon.pdf(x, scale=5), 'r-', lw=2)
axes[1, 2].set_title('Exponential(scale=5)')

plt.tight_layout()
plt.show()
```

```python
# Beta distribution (useful for probabilities)
fig, ax = plt.subplots(figsize=(8, 4))
for a, b in [(0.5, 0.5), (1, 1), (2, 5), (5, 2)]:
    samples = np.random.beta(a, b, n)
    ax.hist(samples, bins=40, alpha=0.5, label=f'Beta({a}, {b})', density=True)
ax.set_title('Beta Distributions')
ax.legend()
plt.show()
```

## Key Observations

- Bernoulli: binary outcomes (success/failure).
- Binomial: count of successes in n independent trials.
- Poisson: count of events in fixed interval (rare events).
- Uniform: equal probability across range.
- Normal: symmetric, bell-shaped, most common in nature.
- Exponential: time between events, memoryless property.

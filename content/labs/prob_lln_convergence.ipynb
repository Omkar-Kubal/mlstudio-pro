# Law of Large Numbers Convergence

Demonstrate that sample means converge to the population mean as sample size increases.

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
```

```python
# Fair coin: E[X] = 0.5 for X ∈ {0, 1}
true_mean = 0.5
max_flips = 10000

flips = np.random.randint(0, 2, max_flips)
cumulative_means = np.cumsum(flips) / np.arange(1, max_flips + 1)

print(f"After 10 flips: mean = {cumulative_means[9]:.3f}")
print(f"After 100 flips: mean = {cumulative_means[99]:.3f}")
print(f"After 1000 flips: mean = {cumulative_means[999]:.3f}")
print(f"After 10000 flips: mean = {cumulative_means[9999]:.3f}")
print(f"True mean: {true_mean}")
```

```python
plt.figure(figsize=(12, 5))
plt.plot(cumulative_means, alpha=0.7)
plt.axhline(true_mean, color='red', linestyle='--', label=f'True mean = {true_mean}')
plt.xscale('log')
plt.xlabel('Number of flips (log scale)')
plt.ylabel('Running average')
plt.title('Law of Large Numbers: Coin Flips')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Multiple sequences to show variability decreasing
n_sequences = 20
max_n = 1000

plt.figure(figsize=(12, 5))
for _ in range(n_sequences):
    flips = np.random.randint(0, 2, max_n)
    running_mean = np.cumsum(flips) / np.arange(1, max_n + 1)
    plt.plot(running_mean, alpha=0.4, linewidth=0.8)

plt.axhline(0.5, color='red', linestyle='--', linewidth=2, label='E[X] = 0.5')
plt.xlabel('Number of trials')
plt.ylabel('Running average')
plt.title('LLN: Multiple Sequences Converging')
plt.legend()
plt.ylim(0.2, 0.8)
plt.show()
```

```python
# Non-uniform distribution: dice roll
true_mean_dice = 3.5
rolls = np.random.randint(1, 7, 5000)
running_mean_dice = np.cumsum(rolls) / np.arange(1, 5001)

plt.figure(figsize=(10, 4))
plt.plot(running_mean_dice)
plt.axhline(true_mean_dice, color='red', linestyle='--', label='E[X] = 3.5')
plt.xlabel('Number of rolls')
plt.ylabel('Running average')
plt.title('LLN: Dice Rolls')
plt.legend()
plt.show()
```

```python
# Deviation from mean decreases as 1/√n
sample_sizes = [10, 50, 100, 500, 1000, 5000]
n_trials = 1000

print("Standard deviation of sample means:")
for n in sample_sizes:
    means = [np.mean(np.random.randint(0, 2, n)) for _ in range(n_trials)]
    std_of_means = np.std(means)
    theoretical = 0.5 / np.sqrt(n)  # σ/√n
    print(f"  n={n:4d}: Observed={std_of_means:.4f}, Theoretical={theoretical:.4f}")
```

## Key Observations

- Sample mean converges to population mean as n → ∞.
- Early samples can deviate significantly from the true mean.
- Convergence rate is proportional to 1/√n.
- Works for any distribution with finite mean and variance.
- Foundation for statistical estimation and inference.

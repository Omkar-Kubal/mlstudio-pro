# Eigenvalues and Eigenvectors

Compute eigenvalues and eigenvectors to understand the natural axes and stretch factors of a transformation.

```python
import numpy as np
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
```

```python
# Simple 2x2 matrix
A = np.array([[3, 1],
              [1, 3]])

eigenvalues, eigenvectors = np.linalg.eig(A)

print(f"Matrix A:\n{A}\n")
print(f"Eigenvalues: {eigenvalues}")
print(f"\nEigenvectors (columns):")
print(eigenvectors)
```

```python
# Verify: A @ v = λ * v
for i in range(len(eigenvalues)):
    lam = eigenvalues[i]
    v = eigenvectors[:, i]
    
    Av = A @ v
    lambda_v = lam * v
    
    print(f"\nEigenvalue λ_{i+1} = {lam:.4f}")
    print(f"  Eigenvector v_{i+1} = {v}")
    print(f"  A @ v = {Av}")
    print(f"  λ * v = {lambda_v}")
    print(f"  Match: {np.allclose(Av, lambda_v)}")
```

```python
# Geometric visualization
fig, ax = plt.subplots(figsize=(8, 8))

# Draw unit circle and its transformation
theta = np.linspace(0, 2*np.pi, 100)
circle = np.array([np.cos(theta), np.sin(theta)])
ellipse = A @ circle

ax.plot(circle[0], circle[1], 'b-', label='Unit circle', alpha=0.5)
ax.plot(ellipse[0], ellipse[1], 'r-', label='Transformed', linewidth=2)

# Plot eigenvectors (scaled by eigenvalues)
for i in range(len(eigenvalues)):
    v = eigenvectors[:, i]
    lam = eigenvalues[i]
    ax.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, 
              color='blue', alpha=0.7, width=0.02)
    ax.quiver(0, 0, lam*v[0], lam*v[1], angles='xy', scale_units='xy', scale=1,
              color='red', alpha=0.7, width=0.02, 
              label=f'λ={lam:.1f}' if i == 0 else None)

ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_title('Eigenvectors define natural axes of transformation')
plt.show()
```

```python
# Matrix power using eigendecomposition
# A^n = V @ Λ^n @ V^(-1)
n = 3

# Direct computation
A_n_direct = np.linalg.matrix_power(A, n)

# Eigendecomposition
V = eigenvectors
Lambda = np.diag(eigenvalues)
V_inv = np.linalg.inv(V)
A_n_eigen = V @ np.linalg.matrix_power(Lambda, n) @ V_inv

print(f"A^{n} (direct):\n{A_n_direct}")
print(f"\nA^{n} (via eigendecomposition):\n{A_n_eigen.real}")
print(f"\nMatch: {np.allclose(A_n_direct, A_n_eigen)}")
```

```python
# Covariance matrix eigenvectors = principal components
np.random.seed(42)
data = np.random.randn(100, 2) @ np.array([[2, 1], [0, 1]])  # Correlated data
cov = np.cov(data.T)

eig_vals, eig_vecs = np.linalg.eig(cov)

print("Covariance matrix eigenvectors (PCA directions):")
print(f"  λ1 = {eig_vals[0]:.4f}, v1 = {eig_vecs[:, 0]}")
print(f"  λ2 = {eig_vals[1]:.4f}, v2 = {eig_vecs[:, 1]}")
print(f"\n  Variance explained: {eig_vals / eig_vals.sum()}")
```

```python
# Properties
print("\nEigenvalue properties:")
print(f"  Sum of eigenvalues = {np.sum(eigenvalues):.4f}")
print(f"  Trace of A = {np.trace(A):.4f}")
print(f"\n  Product of eigenvalues = {np.prod(eigenvalues):.4f}")
print(f"  Determinant of A = {np.linalg.det(A):.4f}")
```

## Key Observations

- Eigenvectors are directions that only get scaled (not rotated) by the matrix.
- Eigenvalues are the scaling factors along those directions.
- trace(A) = sum of eigenvalues; det(A) = product of eigenvalues.
- Covariance matrix eigenvectors = principal component directions.
- Used in PCA, spectral clustering, Google PageRank, and stability analysis.

# Vectors, Scalars, and Operations

Explore vector fundamentals: creation, scalar multiplication, vector addition, and geometric interpretation.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Vector creation
v = np.array([3, 4])
w = np.array([1, 2])

print(f"v = {v}")
print(f"w = {w}")
print(f"v shape: {v.shape}")
```

```python
# Scalar multiplication
scalar = 2
v_scaled = scalar * v

print(f"\nScalar multiplication:")
print(f"  2 * v = {v_scaled}")
print(f"  -1 * v = {-1 * v}")
print(f"  0.5 * v = {0.5 * v}")
```

```python
# Vector addition and subtraction
v_plus_w = v + w
v_minus_w = v - w

print(f"\nVector operations:")
print(f"  v + w = {v_plus_w}")
print(f"  v - w = {v_minus_w}")
```

```python
# Geometric visualization
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

# Scalar multiplication
ax = axes[0]
origin = np.array([0, 0])
ax.quiver(*origin, *v, angles='xy', scale_units='xy', scale=1, color='blue', label='v')
ax.quiver(*origin, *v_scaled, angles='xy', scale_units='xy', scale=1, color='red', alpha=0.7, label='2v')
ax.set_xlim(-1, 8)
ax.set_ylim(-1, 10)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_title('Scalar Multiplication')

# Vector addition (parallelogram rule)
ax = axes[1]
ax.quiver(*origin, *v, angles='xy', scale_units='xy', scale=1, color='blue', label='v')
ax.quiver(*origin, *w, angles='xy', scale_units='xy', scale=1, color='green', label='w')
ax.quiver(*v, *w, angles='xy', scale_units='xy', scale=1, color='green', alpha=0.5)
ax.quiver(*origin, *v_plus_w, angles='xy', scale_units='xy', scale=1, color='red', label='v+w')
ax.set_xlim(-1, 6)
ax.set_ylim(-1, 7)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_title('Vector Addition')

# Vector subtraction
ax = axes[2]
ax.quiver(*origin, *v, angles='xy', scale_units='xy', scale=1, color='blue', label='v')
ax.quiver(*origin, *w, angles='xy', scale_units='xy', scale=1, color='green', label='w')
ax.quiver(*w, *v_minus_w, angles='xy', scale_units='xy', scale=1, color='red', label='v-w')
ax.set_xlim(-1, 5)
ax.set_ylim(-1, 5)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_title('Vector Subtraction (v-w from w to v)')

plt.tight_layout()
plt.show()
```

```python
# Higher dimensions
v_3d = np.array([1, 2, 3])
w_3d = np.array([4, 5, 6])

print(f"\n3D vectors:")
print(f"  v_3d = {v_3d}")
print(f"  w_3d = {w_3d}")
print(f"  v + w = {v_3d + w_3d}")
print(f"  3 * v = {3 * v_3d}")
```

```python
# Linear combination
a, b = 2, 3
linear_combo = a * v + b * w
print(f"\nLinear combination: {a}v + {b}w = {linear_combo}")
```

## Key Observations

- Vectors have magnitude and direction.
- Scalar multiplication scales length (and may reverse direction).
- Vector addition: place tail of second at head of first.
- Subtraction v - w points from w to v.
- Operations extend naturally to any dimension.

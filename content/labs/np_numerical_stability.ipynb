# Numerical Stability

Understand floating-point precision issues and how to avoid them.

```python
import numpy as np
```

```python
# Floating point precision
print(f"0.1 + 0.2 = {0.1 + 0.2}")
print(f"0.1 + 0.2 == 0.3? {0.1 + 0.2 == 0.3}")
print(f"Use np.isclose: {np.isclose(0.1 + 0.2, 0.3)}")
```

```python
# Overflow and underflow
large = np.float64(1e308)
print(f"1e308 * 10 = {large * 10}")  # Overflow → inf

small = np.float64(1e-308)
print(f"1e-308 / 1e10 = {small / 1e10}")  # Underflow → 0
```

```python
# Log-sum-exp trick (softmax stability)
# BAD: overflow for large values
logits = np.array([1000, 1001, 1002])
try:
    bad_softmax = np.exp(logits) / np.sum(np.exp(logits))
    print(f"Naive softmax: {bad_softmax}")
except:
    print("Overflow!")

# GOOD: subtract max for stability
shifted = logits - np.max(logits)
good_softmax = np.exp(shifted) / np.sum(np.exp(shifted))
print(f"Stable softmax: {good_softmax}")
```

```python
# Log probabilities for numerical stability
# BAD: multiply many small probabilities
probs = np.array([0.1, 0.2, 0.05, 0.01])
product = np.prod(probs**100)  # Very small!
print(f"Product of probs^100: {product}")  # Underflow

# GOOD: work in log space
log_probs = np.log(probs)
log_product = np.sum(log_probs * 100)
print(f"Log of product: {log_product}")
```

```python
# Catastrophic cancellation
# BAD: subtracting similar large numbers
a = 1e15 + 1
b = 1e15
print(f"(1e15 + 1) - 1e15 = {a - b}")  # Should be 1.0

# Variance formula: avoid E[X²] - E[X]² for large means
data = np.array([1e8, 1e8 + 1, 1e8 + 2])
var_naive = np.mean(data**2) - np.mean(data)**2
var_stable = np.var(data)
print(f"Naive variance: {var_naive}")
print(f"Stable variance: {var_stable}")
```

## Key Observations
- Never compare floats with ==, use np.isclose
- Subtract max before exp (log-sum-exp trick)
- Work in log space for products of probabilities
- Use stable algorithms from NumPy/SciPy
- Avoid subtracting similar large numbers

# Loss Functions: MSE vs MAE

Compare behavior of MSE and MAE with outliers.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, HuberRegressor
np.random.seed(42)

X = np.linspace(0, 10, 20).reshape(-1, 1)
y = 2*X.ravel() + 3 + np.random.randn(20)*0.5
y[15] = 50  # Outlier
```

```python
# Compare
lr = LinearRegression().fit(X, y)
huber = HuberRegressor().fit(X, y)

plt.figure(figsize=(10, 5))
plt.scatter(X, y, label='Data')
plt.plot(X, lr.predict(X), 'r-', label='OLS (MSE)')
plt.plot(X, huber.predict(X), 'g-', label='Huber')
plt.legend()
plt.title('MSE is sensitive to outliers')
plt.show()

print(f"OLS slope: {lr.coef_[0]:.2f}")
print(f"Huber slope: {huber.coef_[0]:.2f}")
print(f"True slope: 2.0")
```

## Key Observations
- MSE: squares errors, outliers dominate
- MAE: linear penalty, more robust
- Huber: MSE near 0, MAE for large errors

# Conditional Probability Simulation

Simulate conditional probabilities and verify P(A|B) = P(A∩B) / P(B) through frequency counting.

```python
import numpy as np
np.random.seed(42)
```

```python
# Two dice: P(sum >= 10 | first die = 6)
n_trials = 100000
die1 = np.random.randint(1, 7, n_trials)
die2 = np.random.randint(1, 7, n_trials)
total = die1 + die2

# Condition: first die is 6
condition_B = die1 == 6
event_A = total >= 10

# P(A|B) = P(A ∩ B) / P(B)
p_B = np.mean(condition_B)
p_A_and_B = np.mean(event_A & condition_B)
p_A_given_B_formula = p_A_and_B / p_B

# Direct conditional count
p_A_given_B_direct = np.mean(event_A[condition_B])

print("Two dice: P(sum >= 10 | die1 = 6)")
print(f"  P(B) = P(die1=6) = {p_B:.4f}")
print(f"  P(A ∩ B) = {p_A_and_B:.4f}")
print(f"  P(A|B) via formula = {p_A_given_B_formula:.4f}")
print(f"  P(A|B) direct count = {p_A_given_B_direct:.4f}")
print(f"  Expected: 3/6 = 0.5000 (sum>=10 when die1=6 means die2 ∈ {4,5,6})")
```

```python
# Card drawing: P(King | Face card)
# Simulate drawing from a standard deck
deck = list(range(52))  # 0-51
n_draws = 100000

kings = 0
face_cards = 0
kings_given_face = 0

for _ in range(n_draws):
    card = np.random.randint(0, 52)
    rank = card % 13  # 0-12: A, 2, ..., 10, J, Q, K
    
    is_face = rank >= 10  # J, Q, K
    is_king = rank == 12
    
    if is_face:
        face_cards += 1
        if is_king:
            kings_given_face += 1
    if is_king:
        kings += 1

print("\nCards: P(King | Face card)")
print(f"  P(Face) = {face_cards / n_draws:.4f} (expected 12/52 = 0.231)")
print(f"  P(King | Face) = {kings_given_face / face_cards:.4f} (expected 4/12 = 0.333)")
```

```python
# Independence test: two independent events
# A = first coin is heads, B = second coin is heads
n = 100000
coin1 = np.random.choice([0, 1], n)
coin2 = np.random.choice([0, 1], n)

p_A = np.mean(coin1 == 1)
p_B = np.mean(coin2 == 1)
p_A_given_B = np.mean(coin1[coin2 == 1] == 1)

print("\nIndependence check (two coins):")
print(f"  P(A) = {p_A:.4f}")
print(f"  P(A|B) = {p_A_given_B:.4f}")
print(f"  If independent: P(A|B) = P(A) ✓" if abs(p_A - p_A_given_B) < 0.01 else "  Not independent")
```

```python
# Dependent events: drawing without replacement
n_sims = 100000
first_red = 0
second_red_given_first_red = 0

# Bag: 5 red, 5 blue balls
for _ in range(n_sims):
    bag = ['R'] * 5 + ['B'] * 5
    np.random.shuffle(bag)
    
    if bag[0] == 'R':
        first_red += 1
        if bag[1] == 'R':
            second_red_given_first_red += 1

print("\nDependent events (drawing without replacement):")
print(f"  P(first=Red) = {first_red / n_sims:.4f} (expected 0.5)")
print(f"  P(second=Red | first=Red) = {second_red_given_first_red / first_red:.4f} (expected 4/9 = 0.444)")
```

## Key Observations

- P(A|B) = P(A∩B) / P(B) is verified through simulation.
- For independent events: P(A|B) = P(A).
- Drawing without replacement creates dependence.
- Conditional probability reduces the sample space to outcomes where B occurred.

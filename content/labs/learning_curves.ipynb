# Learning Curves

Diagnose under/overfitting from learning curves.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import learning_curve
np.random.seed(42)

X, y = make_classification(n_samples=500, n_features=20, random_state=42)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for ax, model, title in [(axes[0], LogisticRegression(max_iter=1000), 'Logistic (High Bias)'),
                          (axes[1], DecisionTreeClassifier(), 'Tree (High Variance)')]:
    sizes, train, val = learning_curve(model, X, y, train_sizes=np.linspace(0.1, 1, 10), cv=5)
    ax.plot(sizes, train.mean(axis=1), 'b-o', label='Train')
    ax.plot(sizes, val.mean(axis=1), 'r-o', label='Validation')
    ax.set_xlabel('Training Size')
    ax.set_ylabel('Score')
    ax.set_title(title)
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Key Observations
- High bias: both curves plateau low
- High variance: big gap, train high, val low
- More data helps variance, not bias
- Fix bias: more features, less regularization

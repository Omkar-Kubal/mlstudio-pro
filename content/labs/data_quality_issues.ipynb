# Data Quality Issues

Identify and analyze common data quality problems.

```python
import pandas as pd
import numpy as np
np.random.seed(42)

# Simulate messy data
df = pd.DataFrame({
    'name': ['Alice', 'BOB', 'alice', 'Charlie', None, 'Diana'],
    'age': [25, 30, 25, -5, 28, 200],
    'salary': [50000, 60000, 50000, 70000, np.nan, 55000],
    'date': ['2023-01-01', '01/02/2023', '2023-03-01', 'invalid', '2023-05-01', '2023-06-01']
})
print(df)
```

```python
# Missing values
print(f"Missing values:\n{df.isnull().sum()}")

# Duplicates
print(f"\nDuplicate rows: {df.duplicated().sum()}")

# Invalid values
print(f"\nInvalid ages: {df[(df['age'] < 0) | (df['age'] > 120)]['age'].tolist()}")

# Inconsistent formatting
print(f"Unique names: {df['name'].str.lower().dropna().unique()}")

# Mixed date formats
print(f"Date parsing issues: {df['date'].tolist()}")
```

## Key Observations
- Check for: missing, duplicates, invalid ranges, inconsistent formats
- Profile data before modeling

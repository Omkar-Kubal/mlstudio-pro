# BatchNorm and Dropout Effects

Visualize how regularization techniques affect training.

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)

# Batch normalization
def batch_norm(x, eps=1e-5):
    mean = x.mean(axis=0)
    var = x.var(axis=0)
    return (x - mean) / np.sqrt(var + eps)

# Simulate a batch of activations
activations = np.random.randn(100, 10) * 5 + 10  # Mean ~10, high variance
```

```python
normalized = batch_norm(activations)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(activations.ravel(), bins=30, alpha=0.7)
axes[0].set_title(f'Before BatchNorm\nMean: {activations.mean():.2f}, Std: {activations.std():.2f}')
axes[1].hist(normalized.ravel(), bins=30, alpha=0.7)
axes[1].set_title(f'After BatchNorm\nMean: {normalized.mean():.4f}, Std: {normalized.std():.4f}')
plt.tight_layout()
plt.show()
```

```python
# Dropout
def dropout(x, p=0.5):
    mask = np.random.rand(*x.shape) > p
    return x * mask / (1-p)  # Scale to maintain expected value

print("Dropout (p=0.5):")
print(f"Original: {activations[0, :5].round(2)}")
print(f"After:    {dropout(activations, 0.5)[0, :5].round(2)}")
```

## Key Observations
- BatchNorm: normalizes to mean=0, std=1
- Stabilizes training, allows higher learning rates
- Dropout: randomly zeros neurons during training
- Acts as regularization (model averaging)

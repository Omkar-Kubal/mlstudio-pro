# Distribution Plots

Visualize data distributions using histograms, KDEs, and box plots.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(42)
```

```python
# Sample data
normal = np.random.normal(50, 10, 1000)
skewed = np.random.exponential(20, 1000)
bimodal = np.concatenate([np.random.normal(30, 5, 500), np.random.normal(70, 5, 500)])
```

```python
fig, axes = plt.subplots(2, 3, figsize=(14, 8))

# Histograms
for ax, data, title in zip(axes[0], [normal, skewed, bimodal], ['Normal', 'Skewed', 'Bimodal']):
    ax.hist(data, bins=30, edgecolor='black', alpha=0.7)
    ax.set_title(f'{title} - Histogram')

# KDE plots
for ax, data, title in zip(axes[1], [normal, skewed, bimodal], ['Normal', 'Skewed', 'Bimodal']):
    sns.kdeplot(data, ax=ax, fill=True)
    ax.set_title(f'{title} - KDE')

plt.tight_layout()
plt.show()
```

```python
# Box plots and violin plots
import pandas as pd
df = pd.DataFrame({'Normal': normal[:200], 'Skewed': skewed[:200], 'Bimodal': bimodal[:200]})

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
df.boxplot(ax=axes[0])
axes[0].set_title('Box Plots')

df_melted = df.melt(var_name='Distribution', value_name='Value')
sns.violinplot(data=df_melted, x='Distribution', y='Value', ax=axes[1])
axes[1].set_title('Violin Plots')

plt.tight_layout()
plt.show()
```

## Key Observations
- Histograms: see raw counts, bin choice matters
- KDE: smooth estimate of probability density
- Box plots: quartiles, outliers at a glance
- Violin plots: combine box plot + KDE shape

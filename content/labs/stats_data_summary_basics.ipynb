# Data Summary Basics

Compute and compare fundamental summary statistics on synthetic data.
Observe how a single dataset can be characterized by a handful of numbers.

```python
import numpy as np
np.random.seed(42)
```

```python
# Generate synthetic data: exam scores
scores = np.random.normal(loc=72, scale=12, size=200).clip(0, 100)
print(f"Sample size: {len(scores)}")
print(f"First 10 values: {scores[:10].round(1)}")
```

```python
# Core summary statistics
mean = np.mean(scores)
median = np.median(scores)
std = np.std(scores)
min_val, max_val = np.min(scores), np.max(scores)
q25, q75 = np.percentile(scores, [25, 75])
iqr = q75 - q25

print(f"Mean: {mean:.2f}")
print(f"Median: {median:.2f}")
print(f"Std Dev: {std:.2f}")
print(f"Min: {min_val:.2f}, Max: {max_val:.2f}")
print(f"Q25: {q25:.2f}, Q75: {q75:.2f}")
print(f"IQR: {iqr:.2f}")
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(scores, bins=20, edgecolor='black', alpha=0.7)
ax.axvline(mean, color='red', linestyle='--', label=f'Mean={mean:.1f}')
ax.axvline(median, color='green', linestyle='-', label=f'Median={median:.1f}')
ax.set_xlabel('Score')
ax.set_ylabel('Frequency')
ax.set_title('Distribution of Exam Scores')
ax.legend()
plt.tight_layout()
plt.show()
```

## Key Observations

- Mean and median are close when data is roughly symmetric.
- Standard deviation quantifies spread around the mean.
- IQR captures the middle 50% of data, robust to outliers.
- Five numbers (min, Q25, median, Q75, max) summarize distribution shape.

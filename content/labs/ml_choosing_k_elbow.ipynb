# Choosing K: Elbow Method

Use inertia to select optimal number of clusters.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
np.random.seed(42)

X, _ = make_blobs(n_samples=300, centers=4, cluster_std=1.0, random_state=42)
```

```python
inertias = []
K_range = range(1, 10)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X)
    inertias.append(km.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K_range, inertias, 'bo-')
plt.xlabel('K')
plt.ylabel('Inertia (within-cluster sum of squares)')
plt.title('Elbow Method')
plt.axvline(4, color='red', linestyle='--', label='Elbow at K=4')
plt.legend()
plt.show()
```

## Key Observations
- Inertia always decreases with K
- Elbow: point where gain diminishes sharply
- Also consider silhouette score for validation

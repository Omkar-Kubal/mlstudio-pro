# Activation Functions

Visualize common activation functions and their derivatives.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 200)

def sigmoid(x): return 1 / (1 + np.exp(-x))
def tanh(x): return np.tanh(x)
def relu(x): return np.maximum(0, x)
def leaky_relu(x, a=0.1): return np.where(x > 0, x, a*x)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
for ax, (fn, name) in zip(axes.ravel(), [(sigmoid, 'Sigmoid'), (tanh, 'Tanh'), (relu, 'ReLU'), (leaky_relu, 'Leaky ReLU')]):
    ax.plot(x, fn(x), 'b-', linewidth=2)
    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)
    ax.set_title(name)
    ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
print("Output ranges:")
print(f"Sigmoid: (0, 1)")
print(f"Tanh: (-1, 1)")
print(f"ReLU: [0, inf)")
print(f"Leaky ReLU: (-inf, inf)")
print("\nGradient issues:")
print(f"Sigmoid/Tanh: vanishing gradients for |x| >> 0")
print(f"ReLU: dead neurons for x < 0")
print(f"Leaky ReLU: no dead neurons")
```

## Key Observations
- Sigmoid: probabilities, vanishing gradient
- Tanh: zero-centered, still vanishing
- ReLU: fast, sparse, dead neuron problem
- Leaky ReLU / GELU: modern defaults

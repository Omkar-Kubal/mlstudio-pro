# Mean, Median, Mode Effects

Compare how mean, median, and mode behave under different distributions and in the presence of outliers.

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
np.random.seed(42)
```

```python
def compute_central_tendency(data):
    mean = np.mean(data)
    median = np.median(data)
    mode_result = stats.mode(data.round(), keepdims=True)
    mode = mode_result.mode[0]
    return mean, median, mode
```

```python
# Symmetric distribution
symmetric = np.random.normal(50, 10, 1000)
mean, median, mode = compute_central_tendency(symmetric)
print("SYMMETRIC (Normal):")
print(f"  Mean: {mean:.2f}, Median: {median:.2f}, Mode: {mode:.0f}")
```

```python
# Right-skewed distribution
right_skewed = np.random.exponential(20, 1000)
mean, median, mode = compute_central_tendency(right_skewed)
print("\nRIGHT-SKEWED (Exponential):")
print(f"  Mean: {mean:.2f}, Median: {median:.2f}, Mode: {mode:.0f}")
print(f"  Mean > Median (typical for right-skew)")
```

```python
# Left-skewed distribution
left_skewed = 100 - np.random.exponential(20, 1000)
mean, median, mode = compute_central_tendency(left_skewed)
print("\nLEFT-SKEWED:")
print(f"  Mean: {mean:.2f}, Median: {median:.2f}, Mode: {mode:.0f}")
print(f"  Mean < Median (typical for left-skew)")
```

```python
# Effect of outliers on mean vs median
base_data = np.random.normal(50, 5, 100)
outliers = np.array([200, 250, 300])
data_with_outliers = np.concatenate([base_data, outliers])

print("\nOUTLIER EFFECT:")
print(f"  Without outliers - Mean: {np.mean(base_data):.2f}, Median: {np.median(base_data):.2f}")
print(f"  With outliers    - Mean: {np.mean(data_with_outliers):.2f}, Median: {np.median(data_with_outliers):.2f}")
print(f"  Mean shifted by: {np.mean(data_with_outliers) - np.mean(base_data):.2f}")
print(f"  Median shifted by: {np.median(data_with_outliers) - np.median(base_data):.2f}")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

for ax, data, title in zip(axes, 
                           [symmetric, right_skewed, data_with_outliers],
                           ['Symmetric', 'Right-Skewed', 'With Outliers']):
    ax.hist(data, bins=30, alpha=0.7, edgecolor='black')
    ax.axvline(np.mean(data), color='red', linestyle='--', label='Mean')
    ax.axvline(np.median(data), color='green', linestyle='-', label='Median')
    ax.set_title(title)
    ax.legend()

plt.tight_layout()
plt.show()
```

## Key Observations

- Mean = Median for symmetric distributions.
- Mean > Median for right-skewed data (pulled by high values).
- Mean < Median for left-skewed data (pulled by low values).
- Median is robust to outliers; mean is not.
- Mode identifies the most common value, useful for categorical data.

# Random Variables and Sampling

Explore discrete and continuous random variables by sampling and computing properties.

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
```

```python
# Discrete random variable: dice roll
# X ∈ {1, 2, 3, 4, 5, 6}, each with P = 1/6
n_samples = 10000
X_dice = np.random.randint(1, 7, n_samples)

print("Discrete RV (dice):")
print(f"  E[X] = {np.mean(X_dice):.3f} (expected 3.5)")
print(f"  Var(X) = {np.var(X_dice):.3f} (expected 2.917)")
```

```python
# Continuous random variable: normal distribution
# X ~ N(μ=10, σ=3)
mu, sigma = 10, 3
X_normal = np.random.normal(mu, sigma, n_samples)

print("\nContinuous RV (normal):")
print(f"  E[X] = {np.mean(X_normal):.3f} (expected {mu})")
print(f"  Std(X) = {np.std(X_normal):.3f} (expected {sigma})")
```

```python
# Transformation of random variables
# If X ~ N(μ, σ), then Y = 2X + 5 ~ N(2μ + 5, 2σ)
Y = 2 * X_normal + 5

print("\nTransformed RV (Y = 2X + 5):")
print(f"  E[Y] = {np.mean(Y):.3f} (expected {2*mu + 5})")
print(f"  Std(Y) = {np.std(Y):.3f} (expected {2*sigma})")
```

```python
# Sum of random variables
# X1, X2 independent ~ N(μ, σ)
# X1 + X2 ~ N(2μ, σ√2)
X1 = np.random.normal(mu, sigma, n_samples)
X2 = np.random.normal(mu, sigma, n_samples)
Z = X1 + X2

print("\nSum of independent RVs (Z = X1 + X2):")
print(f"  E[Z] = {np.mean(Z):.3f} (expected {2*mu})")
print(f"  Std(Z) = {np.std(Z):.3f} (expected {sigma * np.sqrt(2):.3f})")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

# Discrete
unique, counts = np.unique(X_dice, return_counts=True)
axes[0].bar(unique, counts / n_samples)
axes[0].axhline(1/6, color='red', linestyle='--', label='P=1/6')
axes[0].set_title('Discrete: Dice Roll')
axes[0].legend()

# Continuous
axes[1].hist(X_normal, bins=40, density=True, alpha=0.7, edgecolor='black')
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 100)
axes[1].plot(x, (1/(sigma*np.sqrt(2*np.pi))) * np.exp(-0.5*((x-mu)/sigma)**2), 'r-')
axes[1].set_title('Continuous: Normal(10, 3)')

# Transformed
axes[2].hist(Y, bins=40, density=True, alpha=0.7, edgecolor='black')
axes[2].set_title('Transformed: Y = 2X + 5')

plt.tight_layout()
plt.show()
```

## Key Observations

- Discrete RVs take countable values; continuous RVs have PDF over real line.
- Expected value and variance characterize RV behavior.
- Linear transformation: E[aX + b] = aE[X] + b, Var(aX + b) = a²Var(X).
- Sum of independent RVs: variances add, not standard deviations.

# Gradient Descent from Scratch

Implement gradient descent manually to understand the core optimization algorithm.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def gradient_descent_2d(grad_f, x0, lr=0.1, n_steps=100):
    x = np.array(x0, dtype=float)
    history = [x.copy()]
    for _ in range(n_steps):
        x = x - lr * grad_f(x)
        history.append(x.copy())
    return x, np.array(history)

# f(x, y) = x² + y², grad = [2x, 2y]
grad_f = lambda x: np.array([2*x[0], 2*x[1]])
x_min, hist = gradient_descent_2d(grad_f, [3.0, 2.0], lr=0.1)
print(f"Start: [3.0, 2.0], End: {x_min}")
```

```python
# Visualize
x = np.linspace(-4, 4, 100)
X, Y = np.meshgrid(x, x)
Z = X**2 + Y**2

plt.figure(figsize=(8, 6))
plt.contour(X, Y, Z, levels=20, cmap='viridis')
plt.plot(hist[:, 0], hist[:, 1], 'r.-')
plt.scatter(*hist[0], color='green', s=100, label='Start')
plt.scatter(*hist[-1], color='red', s=100, label='End')
plt.legend()
plt.title('Gradient Descent Path')
plt.show()
```

```python
# Learning rate effects
for lr in [0.01, 0.1, 0.9]:
    _, h = gradient_descent_2d(grad_f, [3.0, 2.0], lr=lr, n_steps=30)
    print(f"lr={lr}: {len(h)} steps to {h[-1].round(4)}")
```

## Key Observations
- Update rule: x ← x - lr × ∇f(x)
- Learning rate too small = slow; too large = unstable
- Foundation of all neural network training

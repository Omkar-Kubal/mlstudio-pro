# Sampling Variability Simulation

Observe how sample statistics vary across repeated samples from the same population. This is the foundation of sampling distributions.

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
```

```python
# Population
population = np.random.exponential(scale=50, size=100000)
true_mean = np.mean(population)
print(f"Population mean: {true_mean:.2f}")
```

```python
# Draw many samples and compute sample means
sample_size = 30
n_samples = 1000

sample_means = []
for _ in range(n_samples):
    sample = np.random.choice(population, size=sample_size, replace=False)
    sample_means.append(np.mean(sample))

sample_means = np.array(sample_means)
print(f"Mean of sample means: {np.mean(sample_means):.2f}")
print(f"Std of sample means: {np.std(sample_means):.2f}")
```

```python
# Theoretical standard error
pop_std = np.std(population)
theoretical_se = pop_std / np.sqrt(sample_size)
print(f"Theoretical SE: {theoretical_se:.2f}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Population distribution
axes[0].hist(population, bins=50, alpha=0.7, edgecolor='black')
axes[0].axvline(true_mean, color='red', linestyle='--')
axes[0].set_title('Population Distribution (Exponential)')
axes[0].set_xlabel('Value')

# Sampling distribution of means
axes[1].hist(sample_means, bins=30, alpha=0.7, edgecolor='black')
axes[1].axvline(true_mean, color='red', linestyle='--', label='True mean')
axes[1].set_title(f'Sampling Distribution of Means (n={sample_size})')
axes[1].set_xlabel('Sample Mean')
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
# Effect of sample size on variability
sample_sizes = [5, 10, 30, 100]
for n in sample_sizes:
    means = [np.mean(np.random.choice(population, n)) for _ in range(500)]
    print(f"n={n:3d}: Std of means = {np.std(means):.2f}")
```

## Key Observations

- Sample means vary around the true population mean.
- The sampling distribution is narrower than the population distribution.
- Larger samples produce less variable sample means (smaller SE).
- SE ≈ σ/√n quantifies this relationship.

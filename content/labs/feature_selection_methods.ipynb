# Feature Selection Methods

Compare filter, wrapper, and embedded feature selection.

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.feature_selection import SelectKBest, f_classif, RFE
from sklearn.linear_model import LogisticRegression
np.random.seed(42)

X, y = make_classification(n_samples=200, n_features=10, n_informative=3, random_state=42)
print(f"Shape: {X.shape}")
```

```python
# Filter method (ANOVA)
selector = SelectKBest(f_classif, k=5)
X_filtered = selector.fit_transform(X, y)
print(f"Filter: selected features {np.where(selector.get_support())[0]}")
print(f"Scores: {selector.scores_.round(2)}")

# Wrapper method (RFE)
rfe = RFE(LogisticRegression(max_iter=1000), n_features_to_select=5)
X_rfe = rfe.fit_transform(X, y)
print(f"\nRFE: selected features {np.where(rfe.support_)[0]}")
```

## Key Observations
- Filter: fast, model-agnostic (correlation, ANOVA)
- Wrapper: uses model performance (RFE, forward/backward)
- Embedded: built into model (Lasso, tree importance)

# Linear Transformations

Visualize how matrices transform vectors geometrically: rotation, scaling, shearing, and reflection.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def plot_transformation(ax, M, title):
    """Plot original and transformed unit square and basis vectors."""
    # Original unit square
    square = np.array([[0, 1, 1, 0, 0],
                       [0, 0, 1, 1, 0]])
    
    # Transform
    transformed = M @ square
    
    ax.plot(square[0], square[1], 'b-', linewidth=2, label='Original')
    ax.plot(transformed[0], transformed[1], 'r-', linewidth=2, label='Transformed')
    
    # Basis vectors
    ax.quiver(0, 0, 1, 0, angles='xy', scale_units='xy', scale=1, color='blue', alpha=0.5)
    ax.quiver(0, 0, 0, 1, angles='xy', scale_units='xy', scale=1, color='blue', alpha=0.5)
    
    # Transformed basis
    ax.quiver(0, 0, M[0, 0], M[1, 0], angles='xy', scale_units='xy', scale=1, color='red', alpha=0.5)
    ax.quiver(0, 0, M[0, 1], M[1, 1], angles='xy', scale_units='xy', scale=1, color='red', alpha=0.5)
    
    ax.set_xlim(-2.5, 2.5)
    ax.set_ylim(-2.5, 2.5)
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.3)
    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)
    ax.set_title(title)
    ax.legend(fontsize=8)
```

```python
# Scaling matrix
scale_x, scale_y = 2, 0.5
M_scale = np.array([[scale_x, 0],
                    [0, scale_y]])

print(f"Scaling matrix (x by {scale_x}, y by {scale_y}):")
print(M_scale)
```

```python
# Rotation matrix
theta = np.radians(45)  # 45 degrees
M_rotate = np.array([[np.cos(theta), -np.sin(theta)],
                     [np.sin(theta), np.cos(theta)]])

print(f"Rotation matrix (45°):")
print(M_rotate.round(3))
```

```python
# Shear matrix
shear = 0.5
M_shear = np.array([[1, shear],
                    [0, 1]])

print(f"Shear matrix (horizontal shear {shear}):")
print(M_shear)
```

```python
# Reflection matrix (about y-axis)
M_reflect = np.array([[-1, 0],
                      [0, 1]])

print(f"Reflection matrix (about y-axis):")
print(M_reflect)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 10))

plot_transformation(axes[0, 0], M_scale, f'Scaling ({scale_x}x, {scale_y}y)')
plot_transformation(axes[0, 1], M_rotate, 'Rotation (45°)')
plot_transformation(axes[1, 0], M_shear, f'Shear (k={shear})')
plot_transformation(axes[1, 1], M_reflect, 'Reflection (y-axis)')

plt.tight_layout()
plt.show()
```

```python
# Composition of transformations
# Rotate then scale = scale then rotate?
v = np.array([1, 0])

rotate_then_scale = M_scale @ M_rotate @ v
scale_then_rotate = M_rotate @ M_scale @ v

print(f"Original: {v}")
print(f"Rotate then scale: {rotate_then_scale.round(3)}")
print(f"Scale then rotate: {scale_then_rotate.round(3)}")
print(f"Order matters: {not np.allclose(rotate_then_scale, scale_then_rotate)}")
```

```python
# Determinant as area scaling
print("\nDeterminants (area scaling factor):")
for name, M in [("Scale", M_scale), ("Rotate", M_rotate), 
                ("Shear", M_shear), ("Reflect", M_reflect)]:
    det = np.linalg.det(M)
    print(f"  {name}: det = {det:.3f}")
print("  (Negative det means orientation flips)")
```

## Key Observations

- Every matrix defines a linear transformation.
- Columns of matrix = where basis vectors land.
- Determinant = area scaling factor (negative = flip).
- Rotation matrices have det = 1, preserve area and angles.
- Transformation order matters: AB ≠ BA geometrically.

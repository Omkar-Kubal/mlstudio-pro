# ROC and PR Curves

Visualize classifier performance across thresholds.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_curve, precision_recall_curve, roc_auc_score, average_precision_score
np.random.seed(42)

X, y = make_classification(n_samples=500, random_state=42)
model = LogisticRegression().fit(X, y)
y_prob = model.predict_proba(X)[:, 1]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# ROC
fpr, tpr, _ = roc_curve(y, y_prob)
axes[0].plot(fpr, tpr, label=f'AUC = {roc_auc_score(y, y_prob):.3f}')
axes[0].plot([0, 1], [0, 1], 'k--')
axes[0].set_xlabel('False Positive Rate')
axes[0].set_ylabel('True Positive Rate')
axes[0].set_title('ROC Curve')
axes[0].legend()

# PR
prec, rec, _ = precision_recall_curve(y, y_prob)
axes[1].plot(rec, prec, label=f'AP = {average_precision_score(y, y_prob):.3f}')
axes[1].set_xlabel('Recall')
axes[1].set_ylabel('Precision')
axes[1].set_title('Precision-Recall Curve')
axes[1].legend()

plt.tight_layout()
plt.show()
```

## Key Observations
- ROC: plots TPR vs FPR at all thresholds
- AUC: area under ROC, 1.0 is perfect, 0.5 is random
- PR curve: better for imbalanced datasets
- Curves above diagonal = better than random

# Matrix Operations

Perform matrix creation, addition, multiplication, and transpose. Understand dimensionality constraints.

```python
import numpy as np
```

```python
# Matrix creation
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[7, 8],
              [9, 10],
              [11, 12]])

print(f"A shape: {A.shape} (2 rows, 3 columns)")
print(f"A:\n{A}\n")

print(f"B shape: {B.shape} (3 rows, 2 columns)")
print(f"B:\n{B}")
```

```python
# Matrix addition (same shape required)
C = np.array([[1, 1, 1],
              [2, 2, 2]])

print("Matrix addition A + C:")
print(A + C)
```

```python
# Matrix multiplication
# A (2×3) × B (3×2) = result (2×2)
AB = A @ B  # or np.matmul(A, B) or np.dot(A, B)

print(f"A @ B shape: {AB.shape}")
print(f"A @ B:\n{AB}")

# Manual calculation for (0,0) element
manual_00 = 1*7 + 2*9 + 3*11
print(f"\nManual (0,0): 1*7 + 2*9 + 3*11 = {manual_00}")
```

```python
# Order matters! AB ≠ BA (when both are defined)
# B @ A is (3×2) × (2×3) = (3×3)
BA = B @ A

print(f"B @ A shape: {BA.shape}")
print(f"B @ A:\n{BA}")
print(f"\nAB ≠ BA: Different shapes!")
```

```python
# Transpose
print(f"A:\n{A}")
print(f"\nA.T (transpose):\n{A.T}")
print(f"\nOriginal shape: {A.shape}, Transposed shape: {A.T.shape}")
```

```python
# Special matrices
I = np.eye(3)  # Identity matrix
Z = np.zeros((2, 3))  # Zero matrix
O = np.ones((2, 2))  # Ones matrix

print("Identity (3×3):")
print(I)

# Identity property: A @ I = A
print(f"\nA @ I = A? {np.allclose(A @ np.eye(3), A)}")
```

```python
# Matrix-vector multiplication
M = np.array([[2, 0],
              [0, 3]])
v = np.array([1, 1])

result = M @ v
print(f"Matrix-vector multiplication:")
print(f"M @ v = {result}")
print(f"(Scales x by 2, y by 3)")
```

```python
# Element-wise vs matrix multiplication
D = np.array([[1, 2],
              [3, 4]])
E = np.array([[5, 6],
              [7, 8]])

print("Element-wise (D * E):")
print(D * E)

print("\nMatrix multiplication (D @ E):")
print(D @ E)
```

## Key Observations

- Matrix multiplication requires matching inner dimensions: (m×n) @ (n×p) = (m×p).
- Matrix multiplication is NOT commutative: AB ≠ BA in general.
- Transpose swaps rows and columns: (A.T).T = A.
- Identity matrix is the multiplicative neutral element.
- Element-wise (*) and matrix (@) multiplication are different!

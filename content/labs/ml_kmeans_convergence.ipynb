# K-Means Convergence

Visualize K-Means clustering iterations.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
np.random.seed(42)

X, _ = make_blobs(n_samples=300, centers=3, cluster_std=1.0, random_state=42)
```

```python
fig, axes = plt.subplots(2, 3, figsize=(12, 8))
for i, ax in enumerate(axes.ravel()):
    km = KMeans(n_clusters=3, init='random', n_init=1, max_iter=i+1, random_state=42)
    labels = km.fit_predict(X)
    ax.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', alpha=0.6)
    ax.scatter(km.cluster_centers_[:, 0], km.cluster_centers_[:, 1], c='red', marker='X', s=200)
    ax.set_title(f'Iteration {i+1}')
plt.tight_layout()
plt.show()
```

## Key Observations
- K-Means alternates: assign points, update centroids
- Converges when assignments stop changing
- Random init can lead to different results

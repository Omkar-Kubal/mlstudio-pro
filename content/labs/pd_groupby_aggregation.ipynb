# GroupBy and Aggregation

Use Pandas groupby for split-apply-combine operations.

```python
import pandas as pd
import numpy as np
np.random.seed(42)

df = pd.DataFrame({
    'department': np.random.choice(['Sales', 'IT', 'HR'], 100),
    'region': np.random.choice(['East', 'West'], 100),
    'salary': np.random.randint(40000, 90000, 100),
    'experience': np.random.randint(1, 15, 100)
})
print(df.head())
```

```python
# Basic groupby
dept_stats = df.groupby('department')['salary'].mean()
print(f"Mean salary by department:\n{dept_stats}")
```

```python
# Multiple aggregations
agg = df.groupby('department').agg({
    'salary': ['mean', 'min', 'max', 'count'],
    'experience': 'mean'
})
print(agg)
```

```python
# Multi-level groupby
multi = df.groupby(['department', 'region'])['salary'].mean().unstack()
print(multi)
```

```python
# Transform vs aggregate
# Aggregate: reduces to one value per group
# Transform: returns same-size output
df['dept_mean_salary'] = df.groupby('department')['salary'].transform('mean')
df['salary_vs_dept'] = df['salary'] - df['dept_mean_salary']
print(df[['department', 'salary', 'dept_mean_salary', 'salary_vs_dept']].head())
```

## Key Observations
- groupby splits data by key, applies function, combines results
- agg() for multiple functions on multiple columns
- transform() keeps original shape (for comparisons)
- unstack() pivots multi-level index to columns

# Text as Vectors

Convert text to numerical representations.

```python
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

docs = [
    'machine learning is amazing',
    'deep learning is a subset of machine learning',
    'natural language processing uses machine learning'
]
```

```python
# Bag of Words
cv = CountVectorizer()
bow = cv.fit_transform(docs)
print(f"Vocabulary: {cv.get_feature_names_out()}")
print(f"BoW matrix:\n{bow.toarray()}")
```

```python
# TF-IDF
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(docs)
print(f"\nTF-IDF matrix (rounded):\n{tfidf_matrix.toarray().round(2)}")
```

```python
# Document similarity
from sklearn.metrics.pairwise import cosine_similarity
sim = cosine_similarity(tfidf_matrix)
print(f"\nDocument similarity:\n{sim.round(2)}")
```

## Key Observations
- Bag of Words: word counts, ignores order
- TF-IDF: downweights common words
- Sparse high-dimensional vectors
- Word embeddings (Word2Vec, BERT) capture semantics

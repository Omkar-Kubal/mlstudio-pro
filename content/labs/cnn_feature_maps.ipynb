# Feature Maps

Visualize how convolutional layers extract hierarchical features.

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulate an image with patterns
np.random.seed(42)
image = np.zeros((28, 28))
image[10:20, 5:10] = 1   # Vertical line
image[5:10, 10:20] = 1   # Horizontal line
image[18:25, 15:25] = 0.5  # Square
```

```python
# Multiple learned filters
filters = [
    np.array([[-1, 0, 1], [-1, 0, 1], [-1, 0, 1]]),  # Vertical
    np.array([[-1, -1, -1], [0, 0, 0], [1, 1, 1]]),  # Horizontal
    np.array([[-1, -1, -1], [-1, 8, -1], [-1, -1, -1]])  # Edge
]

def convolve(img, k):
    h, w = img.shape
    kh, kw = k.shape
    out = np.zeros((h-kh+1, w-kw+1))
    for i in range(out.shape[0]):
        for j in range(out.shape[1]):
            out[i,j] = np.sum(img[i:i+kh, j:j+kw] * k)
    return np.maximum(0, out)  # ReLU

fig, axes = plt.subplots(1, 4, figsize=(14, 3))
axes[0].imshow(image, cmap='gray')
axes[0].set_title('Input')
for ax, f, title in zip(axes[1:], filters, ['Vertical', 'Horizontal', 'Edge']):
    ax.imshow(convolve(image, f), cmap='viridis')
    ax.set_title(f'{title} Filter')
    ax.axis('off')
plt.show()
```

## Key Observations
- Each filter produces one feature map
- Deep layers: low-level → high-level features
- ReLU after convolution introduces non-linearity
- CNN learns filters automatically from data

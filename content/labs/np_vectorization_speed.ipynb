# Vectorization Speed

Compare vectorized NumPy operations vs Python loops to understand performance gains.

```python
import numpy as np
import time
```

```python
# Setup
n = 1_000_000
a = np.random.randn(n)
b = np.random.randn(n)
a_list = a.tolist()
b_list = b.tolist()
```

```python
# Element-wise addition: loop vs vectorized
# Python loop
start = time.time()
result_loop = [a_list[i] + b_list[i] for i in range(n)]
loop_time = time.time() - start

# NumPy vectorized
start = time.time()
result_numpy = a + b
numpy_time = time.time() - start

print(f"Addition of {n:,} elements:")
print(f"  Python loop: {loop_time:.4f}s")
print(f"  NumPy:       {numpy_time:.4f}s")
print(f"  Speedup:     {loop_time/numpy_time:.0f}x")
```

```python
# Dot product
start = time.time()
dot_loop = sum(a_list[i] * b_list[i] for i in range(n))
loop_time = time.time() - start

start = time.time()
dot_numpy = np.dot(a, b)
numpy_time = time.time() - start

print(f"\nDot product:")
print(f"  Python loop: {loop_time:.4f}s")
print(f"  NumPy:       {numpy_time:.6f}s")
print(f"  Speedup:     {loop_time/numpy_time:.0f}x")
```

```python
# Matrix multiplication
m = 500
A = np.random.randn(m, m)
B = np.random.randn(m, m)

# Pure Python (O(n³))
start = time.time()
C_loop = [[sum(A[i,k]*B[k,j] for k in range(m)) for j in range(m)] for i in range(m)]
loop_time = time.time() - start

# NumPy
start = time.time()
C_numpy = A @ B
numpy_time = time.time() - start

print(f"\n{m}x{m} matrix multiplication:")
print(f"  Python loop: {loop_time:.2f}s")
print(f"  NumPy:       {numpy_time:.4f}s")
print(f"  Speedup:     {loop_time/numpy_time:.0f}x")
```

```python
# Why? NumPy uses:
# - C/Fortran compiled code
# - BLAS/LAPACK libraries
# - SIMD instructions (CPU vectorization)
# - Cache-friendly memory access
print("\nNumPy config:")
np.show_config()
```

## Key Observations
- Vectorized NumPy is 10-1000x faster than Python loops
- Speedup comes from compiled code and hardware optimization
- Always use broadcasting and ufuncs over explicit loops
- The gap grows with data size

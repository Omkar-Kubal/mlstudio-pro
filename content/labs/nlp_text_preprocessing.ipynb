# Text Preprocessing Pipeline

Standardize text for NLP models.

```python
import re

text = "Hello World! This is an EXAMPLE of Text Processing... #NLP @2024"
```

```python
def preprocess(text):
    # Lowercase
    text = text.lower()
    # Remove special characters
    text = re.sub(r'[^a-z\s]', '', text)
    # Remove extra whitespace
    text = ' '.join(text.split())
    return text

print(f"Original: {text}")
print(f"Processed: {preprocess(text)}")
```

```python
# Tokenization
from sklearn.feature_extraction.text import CountVectorizer

docs = ['Machine learning models', 'Deep learning networks', 'Neural network training']
cv = CountVectorizer(ngram_range=(1, 2))  # Unigrams and bigrams
X = cv.fit_transform(docs)

print(f"Tokens: {cv.get_feature_names_out()}")
```

## Key Observations
- Lowercase for consistency
- Remove punctuation, numbers (domain-specific)
- Tokenization splits text into units
- N-grams capture word sequences

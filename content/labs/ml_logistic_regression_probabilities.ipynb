# Logistic Regression Probabilities

Understand sigmoid output as probability.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
np.random.seed(42)

X = np.random.randn(100, 1)
y = (X.ravel() > 0).astype(int)

model = LogisticRegression().fit(X, y)
```

```python
X_test = np.linspace(-3, 3, 100).reshape(-1, 1)
probs = model.predict_proba(X_test)[:, 1]

plt.figure(figsize=(10, 5))
plt.scatter(X, y, alpha=0.5, label='Data')
plt.plot(X_test, probs, 'r-', linewidth=2, label='P(y=1)')
plt.axhline(0.5, color='green', linestyle='--', label='Threshold 0.5')
plt.xlabel('X')
plt.ylabel('Probability')
plt.legend()
plt.title('Logistic Regression: Sigmoid Probability')
plt.show()

print(f"P(y=1 | x=-2): {model.predict_proba([[-2]])[0, 1]:.3f}")
print(f"P(y=1 | x=0):  {model.predict_proba([[0]])[0, 1]:.3f}")
print(f"P(y=1 | x=2):  {model.predict_proba([[2]])[0, 1]:.3f}")
```

## Key Observations
- Logistic regression outputs probabilities via sigmoid
- predict_proba() gives [P(y=0), P(y=1)]
- Default threshold 0.5, can adjust for precision/recall tradeoff

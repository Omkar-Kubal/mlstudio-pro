# Convex vs Non-Convex Functions

Understand convexity and its implications for optimization guarantees.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-3, 3, 200)

# Convex functions
f_quadratic = x ** 2
f_exponential = np.exp(x)
f_abs = np.abs(x)

# Non-convex functions
f_sine = np.sin(2 * x)
f_double_well = (x**2 - 1)**2
```

```python
fig, axes = plt.subplots(2, 3, figsize=(14, 8))

# Convex
for ax, (f, title) in zip(axes[0], [
    (f_quadratic, 'x² (Convex)'),
    (f_exponential, 'exp(x) (Convex)'),
    (f_abs, '|x| (Convex)')
]):
    ax.plot(x, f, 'b-', linewidth=2)
    ax.set_title(title)
    ax.grid(True, alpha=0.3)

# Non-convex
for ax, (f, title) in zip(axes[1], [
    (f_sine, 'sin(2x) (Non-convex)'),
    (f_double_well, '(x²-1)² (Non-convex)'),
    (None, 'Convexity Test')
]):
    if f is not None:
        ax.plot(x, f, 'r-', linewidth=2)
        ax.set_title(title)
        ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Convexity test: f(λa + (1-λ)b) ≤ λf(a) + (1-λ)f(b)
# Geometrically: line segment between any two points lies above the curve

def test_convexity(f, x, name, n_tests=100):
    violations = 0
    for _ in range(n_tests):
        a, b = np.random.choice(x, 2)
        lam = np.random.uniform(0, 1)
        
        midpoint = lam * a + (1 - lam) * b
        f_midpoint = f(midpoint)
        f_chord = lam * f(a) + (1 - lam) * f(b)
        
        if f_midpoint > f_chord + 1e-10:  # Small tolerance
            violations += 1
    
    return violations

print("Convexity test (violations in 100 random tests):")
print(f"  x²:       {test_convexity(lambda x: x**2, x, 'quadratic')} (convex)")
print(f"  exp(x):   {test_convexity(np.exp, x, 'exp')} (convex)")
print(f"  sin(2x):  {test_convexity(lambda x: np.sin(2*x), x, 'sine')} (non-convex)")
print(f"  (x²-1)²:  {test_convexity(lambda x: (x**2-1)**2, x, 'double_well')} (non-convex)")
```

```python
# Visual demonstration of convexity definition
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

x_range = np.linspace(-2, 2, 100)

# Convex example
ax = axes[0]
f_conv = lambda x: x**2
ax.plot(x_range, f_conv(x_range), 'b-', linewidth=2)

a, b = -1.5, 1.0
ax.plot([a, b], [f_conv(a), f_conv(b)], 'r-', linewidth=2, label='Chord')
ax.scatter([a, b], [f_conv(a), f_conv(b)], color='red', s=100, zorder=5)

lam = 0.4
mid = lam * a + (1-lam) * b
ax.scatter([mid], [f_conv(mid)], color='green', s=100, zorder=5, label='f(midpoint)')
ax.scatter([mid], [lam*f_conv(a) + (1-lam)*f_conv(b)], color='orange', s=100, 
           marker='s', zorder=5, label='Chord at midpoint')
ax.set_title('Convex: Curve below chord')
ax.legend()
ax.grid(True, alpha=0.3)

# Non-convex example
ax = axes[1]
f_nonconv = lambda x: np.sin(2*x) + 0.2*x**2
ax.plot(x_range, f_nonconv(x_range), 'b-', linewidth=2)

a, b = -1.5, 1.5
ax.plot([a, b], [f_nonconv(a), f_nonconv(b)], 'r-', linewidth=2, label='Chord')
ax.scatter([a, b], [f_nonconv(a), f_nonconv(b)], color='red', s=100, zorder=5)

mid = 0.5 * a + 0.5 * b
ax.scatter([mid], [f_nonconv(mid)], color='green', s=100, zorder=5, label='f(midpoint)')
ax.set_title('Non-convex: Curve can be above chord')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Implications for optimization
print("\nOptimization implications:")
print("\nCONVEX FUNCTIONS:")
print("  ✓ Any local minimum is THE global minimum")
print("  ✓ Gradient descent always converges to optimal")
print("  ✓ Linear regression loss is convex")
print("\nNON-CONVEX FUNCTIONS:")
print("  ✗ Multiple local minima possible")
print("  ✗ Optimization can get stuck")
print("  ✗ Neural network losses are non-convex")
print("  → Initialization matters, may need restarts")
```

## Key Observations

- Convex: line segment between two points lies above the curve.
- Convex functions have a single global minimum.
- Linear regression, logistic regression have convex losses.
- Neural networks have non-convex losses (many local minima).
- Convexity provides optimization guarantees; non-convexity requires care.

# Backpropagation Numerical

Verify gradients numerically for a simple neural network.

```python
import numpy as np
np.random.seed(42)

def sigmoid(x): return 1 / (1 + np.exp(-x))
def sigmoid_grad(x): s = sigmoid(x); return s * (1 - s)

# Simple network: 2 inputs, 2 hidden, 1 output
X = np.array([[0.5, 0.3]])
y = np.array([[0.8]])
W1 = np.random.randn(2, 2) * 0.5
W2 = np.random.randn(2, 1) * 0.5
```

```python
# Forward pass
z1 = X @ W1
a1 = sigmoid(z1)
z2 = a1 @ W2
a2 = sigmoid(z2)
loss = 0.5 * (y - a2)**2
print(f"Loss: {loss[0,0]:.6f}")

# Backward pass (analytical gradients)
d2 = -(y - a2) * sigmoid_grad(z2)
dW2 = a1.T @ d2
d1 = (d2 @ W2.T) * sigmoid_grad(z1)
dW1 = X.T @ d1
print(f"\nAnalytical dW2[0,0]: {dW2[0,0]:.6f}")
```

```python
# Numerical gradient verification
def compute_loss(W1, W2, X, y):
    a1 = sigmoid(X @ W1)
    a2 = sigmoid(a1 @ W2)
    return 0.5 * (y - a2)**2

eps = 1e-5
W2_plus = W2.copy(); W2_plus[0,0] += eps
W2_minus = W2.copy(); W2_minus[0,0] -= eps
numerical_grad = (compute_loss(W1, W2_plus, X, y) - compute_loss(W1, W2_minus, X, y)) / (2*eps)
print(f"Numerical dW2[0,0]: {numerical_grad[0,0]:.6f}")
print(f"Difference: {abs(dW2[0,0] - numerical_grad[0,0]):.10f}")
```

## Key Observations
- Backprop: chain rule applied layer by layer
- Numerical gradient: (L(w+ε) - L(w-ε)) / 2ε
- Use numerical check to verify implementations
- Differences < 1e-7 indicate correct gradients

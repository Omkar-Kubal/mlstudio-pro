# Convolution Operation

Implement and visualize 2D convolution from scratch.

```python
import numpy as np
import matplotlib.pyplot as plt

def convolve2d(image, kernel):
    kh, kw = kernel.shape
    h, w = image.shape
    out = np.zeros((h-kh+1, w-kw+1))
    for i in range(out.shape[0]):
        for j in range(out.shape[1]):
            out[i,j] = np.sum(image[i:i+kh, j:j+kw] * kernel)
    return out
```

```python
# Sample image (simple gradient)
image = np.zeros((10, 10))
image[3:7, 3:7] = 1  # White square

# Edge detection kernels
sobel_x = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]])
sobel_y = np.array([[-1, -2, -1], [0, 0, 0], [1, 2, 1]])

edge_x = convolve2d(image, sobel_x)
edge_y = convolve2d(image, sobel_y)
edge = np.sqrt(edge_x**2 + edge_y**2)

fig, axes = plt.subplots(1, 4, figsize=(14, 3))
for ax, img, title in zip(axes, [image, edge_x, edge_y, edge], ['Original', 'Sobel X', 'Sobel Y', 'Edge Magnitude']):
    ax.imshow(img, cmap='gray')
    ax.set_title(title)
    ax.axis('off')
plt.show()
```

## Key Observations
- Convolution: sliding window, element-wise multiply, sum
- Different kernels detect different features
- Output size: (H-kH+1, W-kW+1) without padding
- CNNs learn optimal kernels from data

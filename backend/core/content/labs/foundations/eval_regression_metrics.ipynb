# Regression Metrics

Compute and compare MSE, RMSE, MAE, and R² for regression.

```python
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
np.random.seed(42)

y_true = np.array([3, 5, 2, 7, 5, 4, 6, 3])
y_pred = np.array([2.5, 5, 2.5, 7.5, 4.5, 4, 6.5, 3])
```

```python
mse = mean_squared_error(y_true, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_true, y_pred)
r2 = r2_score(y_true, y_pred)

print(f"MSE:  {mse:.4f} (penalizes large errors more)")
print(f"RMSE: {rmse:.4f} (same units as y)")
print(f"MAE:  {mae:.4f} (average absolute error)")
print(f"R²:   {r2:.4f} (explained variance, 1 is perfect)")
```

```python
# Manual calculations
print(f"\nManual MSE: {np.mean((y_true - y_pred)**2):.4f}")
print(f"Manual MAE: {np.mean(np.abs(y_true - y_pred)):.4f}")
ss_res = np.sum((y_true - y_pred)**2)
ss_tot = np.sum((y_true - np.mean(y_true))**2)
print(f"Manual R²: {1 - ss_res/ss_tot:.4f}")
```

## Key Observations
- MSE: squared errors, sensitive to outliers
- RMSE: interpretable in original units
- MAE: robust to outliers
- R²: 0-1 (can be negative for bad models)

# Perceptron from Scratch

Implement a single-layer perceptron for binary classification.

```python
import numpy as np
np.random.seed(42)

# Linearly separable data
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0, 0, 0, 1])  # AND gate
```

```python
def perceptron(X, y, lr=0.1, epochs=100):
    n_samples, n_features = X.shape
    w = np.zeros(n_features)
    b = 0
    
    for _ in range(epochs):
        for i in range(n_samples):
            z = np.dot(X[i], w) + b
            y_pred = 1 if z >= 0 else 0
            update = lr * (y[i] - y_pred)
            w += update * X[i]
            b += update
    return w, b

w, b = perceptron(X, y)
print(f"Weights: {w}, Bias: {b}")
print(f"Predictions: {[1 if np.dot(x, w)+b >= 0 else 0 for x in X]}")
print(f"True: {y.tolist()}")
```

## Key Observations
- Perceptron: simplest neural network, one neuron
- Update: w += lr * (y - y_pred) * x
- Only converges for linearly separable data
- Cannot solve XOR (non-linear)

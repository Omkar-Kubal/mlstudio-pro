# Gradient Visualization

Compute and visualize gradients as vectors pointing in the direction of steepest ascent.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1D gradient visualization
x = np.linspace(-3, 3, 100)
f = x ** 2
gradient = 2 * x  # df/dx = 2x

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x, f, 'b-', linewidth=2, label='f(x) = x²')
ax.plot(x, gradient, 'r--', linewidth=2, label="f'(x) = 2x")

# Show tangent lines at specific points
for x0 in [-2, -1, 0, 1, 2]:
    slope = 2 * x0
    y0 = x0 ** 2
    tangent = slope * (x - x0) + y0
    ax.plot(x, tangent, 'g-', alpha=0.3, linewidth=1)
    ax.scatter([x0], [y0], color='green', s=50, zorder=5)

ax.set_xlabel('x')
ax.set_ylim(-5, 10)
ax.legend()
ax.set_title('Function and its gradient (slope of tangent)')
ax.grid(True, alpha=0.3)
plt.show()
```

```python
# 2D gradient field
# f(x, y) = x² + y²
# ∇f = [2x, 2y]

x = np.linspace(-2, 2, 20)
y = np.linspace(-2, 2, 20)
X, Y = np.meshgrid(x, y)

# Function values for contours
Z = X**2 + Y**2

# Gradients
dZ_dx = 2 * X
dZ_dy = 2 * Y

plt.figure(figsize=(10, 8))
plt.contour(X, Y, Z, levels=15, cmap='viridis', alpha=0.5)
plt.quiver(X, Y, dZ_dx, dZ_dy, color='red', alpha=0.7)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gradient Field: ∇f points toward steepest ascent\n(Negative gradient = steepest descent)')
plt.colorbar(label='f(x,y) = x² + y²')
plt.axis('equal')
plt.show()
```

```python
# Gradient magnitude = steepness
gradient_magnitude = np.sqrt(dZ_dx**2 + dZ_dy**2)

plt.figure(figsize=(10, 8))
plt.contourf(X, Y, gradient_magnitude, levels=20, cmap='hot')
plt.colorbar(label='||∇f||')
plt.contour(X, Y, Z, levels=10, colors='white', alpha=0.5)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gradient Magnitude (steepness of function)')
plt.axis('equal')
plt.show()
```

```python
# Numerical gradient computation
def numerical_gradient(f, x, h=1e-5):
    """Compute gradient numerically using finite differences."""
    grad = np.zeros_like(x)
    for i in range(len(x)):
        x_plus = x.copy()
        x_minus = x.copy()
        x_plus[i] += h
        x_minus[i] -= h
        grad[i] = (f(x_plus) - f(x_minus)) / (2 * h)
    return grad

# Test on f(x, y) = x² + y²
f = lambda x: x[0]**2 + x[1]**2
point = np.array([1.0, 2.0])

numerical = numerical_gradient(f, point)
analytical = np.array([2 * point[0], 2 * point[1]])

print(f"Point: {point}")
print(f"Numerical gradient: {numerical}")
print(f"Analytical gradient: {analytical}")
print(f"Match: {np.allclose(numerical, analytical)}")
```

```python
# Gradient perpendicular to level curves
x = np.linspace(-2, 2, 30)
y = np.linspace(-2, 2, 30)
X, Y = np.meshgrid(x, y)

# Ellipse: f(x,y) = x² + 4y²
Z_ellipse = X**2 + 4*Y**2
dZ_dx = 2 * X
dZ_dy = 8 * Y

# Normalize for visualization
magnitude = np.sqrt(dZ_dx**2 + dZ_dy**2) + 1e-10
dZ_dx_norm = dZ_dx / magnitude
dZ_dy_norm = dZ_dy / magnitude

plt.figure(figsize=(10, 8))
plt.contour(X, Y, Z_ellipse, levels=15, cmap='viridis')
plt.quiver(X[::2, ::2], Y[::2, ::2], dZ_dx_norm[::2, ::2], dZ_dy_norm[::2, ::2], 
           color='red', alpha=0.7)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gradient is perpendicular to level curves')
plt.axis('equal')
plt.show()
```

## Key Observations

- Gradient ∇f is a vector pointing in direction of steepest ascent.
- |∇f| gives the rate of change (steepness) in that direction.
- -∇f points toward steepest descent (used in optimization).
- Gradient is perpendicular to level curves (contour lines).
- Numerical gradients can approximate analytical when formula unavailable.

# Python Data Types and Structures

Explore Python's core data types and structures for data science.

```python
# Numeric types
integer = 42
floating = 3.14159
complex_num = 2 + 3j

print(f"Types: {type(integer)}, {type(floating)}, {type(complex_num)}")
print(f"Integer division: 7 // 3 = {7 // 3}")
print(f"Modulo: 7 % 3 = {7 % 3}")
print(f"Power: 2 ** 10 = {2 ** 10}")
```

```python
# Lists (mutable, ordered)
data = [1, 2, 3, 4, 5]
data.append(6)
data.extend([7, 8])

print(f"List: {data}")
print(f"Slicing: {data[2:5]}, {data[::2]}, {data[::-1]}")
print(f"List comprehension: {[x**2 for x in data if x % 2 == 0]}")
```

```python
# Dictionaries (key-value mapping)
model_params = {'learning_rate': 0.01, 'epochs': 100, 'batch_size': 32}

print(f"Access: {model_params['learning_rate']}")
print(f"Keys: {list(model_params.keys())}")
print(f"Items: {list(model_params.items())}")

# Dict comprehension
squared = {x: x**2 for x in range(5)}
print(f"Dict comp: {squared}")
```

```python
# Sets (unique, unordered)
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(f"Union: {a | b}")
print(f"Intersection: {a & b}")
print(f"Difference: {a - b}")
```

```python
# Tuples (immutable)
point = (3, 4)
x, y = point  # Unpacking
print(f"Point: {point}, x={x}, y={y}")

# Named tuples for clarity
from collections import namedtuple
Point = namedtuple('Point', ['x', 'y'])
p = Point(3, 4)
print(f"Named tuple: {p.x}, {p.y}")
```

## Key Observations
- Lists: mutable, ordered, use for sequences
- Dicts: fast lookup by key, use for mappings
- Sets: unique elements, fast membership testing
- Tuples: immutable, use for fixed structures

# Linear Algebra with NumPy

Use NumPy's linalg module for common linear algebra operations.

```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)
```

```python
# Matrix multiplication
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print(f"A @ B:\n{A @ B}")
print(f"\nnp.matmul(A, B):\n{np.matmul(A, B)}")
```

```python
# Inverse and solve
A_inv = np.linalg.inv(A)
print(f"A inverse:\n{A_inv}")
print(f"A @ A_inv:\n{A @ A_inv}")

# Solve Ax = b
b = np.array([5, 11])
x = np.linalg.solve(A, b)
print(f"\nSolve Ax = b: x = {x}")
print(f"Verify: A @ x = {A @ x}")
```

```python
# Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)
print(f"Eigenvalues: {eigenvalues}")
print(f"Eigenvectors:\n{eigenvectors}")
```

```python
# SVD
M = np.array([[1, 2, 3], [4, 5, 6]])
U, S, Vt = np.linalg.svd(M)

print(f"M shape: {M.shape}")
print(f"U shape: {U.shape}")
print(f"S: {S}")
print(f"Vt shape: {Vt.shape}")

# Reconstruct
S_full = np.zeros((2, 3))
np.fill_diagonal(S_full, S)
M_reconstructed = U @ S_full @ Vt
print(f"\nReconstructed M:\n{M_reconstructed}")
```

```python
# Norms and determinant
v = np.array([3, 4])
print(f"L2 norm: {np.linalg.norm(v)}")
print(f"L1 norm: {np.linalg.norm(v, ord=1)}")
print(f"det(A): {np.linalg.det(A)}")
print(f"rank(M): {np.linalg.matrix_rank(M)}")
```

## Key Observations
- Use @ or np.matmul for matrix multiplication
- np.linalg.solve is more stable than computing inverse
- SVD is the foundation of PCA and matrix compression
- NumPy uses optimized BLAS/LAPACK libraries

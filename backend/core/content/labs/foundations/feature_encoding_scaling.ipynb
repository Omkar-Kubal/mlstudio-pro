# Feature Encoding and Scaling

Transform categorical and numerical features for ML.

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler, MinMaxScaler

df = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'red', 'blue'],
    'size': ['S', 'M', 'L', 'M', 'S'],
    'price': [10, 20, 30, 15, 25]
})
print(df)
```

```python
# Label encoding (ordinal)
le = LabelEncoder()
df['size_encoded'] = le.fit_transform(df['size'])
print(f"Size encoded: {dict(zip(le.classes_, range(len(le.classes_))))}")

# One-hot encoding (nominal)
color_onehot = pd.get_dummies(df['color'], prefix='color')
print(f"\nOne-hot:\n{color_onehot}")
```

```python
# Scaling
scaler_std = StandardScaler()
scaler_mm = MinMaxScaler()

price = df[['price']]
print(f"Original: {price['price'].tolist()}")
print(f"StandardScaler: {scaler_std.fit_transform(price).ravel().round(2)}")
print(f"MinMaxScaler: {scaler_mm.fit_transform(price).ravel().round(2)}")
```

## Key Observations
- Label encoding for ordinal categories
- One-hot for nominal (no order)
- StandardScaler: mean=0, std=1
- MinMaxScaler: range [0, 1]

# Bayes' Theorem: Medical Test Example

Simulate Bayes' theorem in a classic medical testing scenario to understand posterior probability.

```python
import numpy as np
np.random.seed(42)
```

```python
# Medical test parameters
prevalence = 0.01        # P(Disease) = 1% have the disease
sensitivity = 0.95       # P(Positive | Disease) = 95%
specificity = 0.90       # P(Negative | No Disease) = 90%
false_positive_rate = 1 - specificity  # P(Positive | No Disease) = 10%

print("Medical test parameters:")
print(f"  Prevalence: {prevalence:.1%}")
print(f"  Sensitivity: {sensitivity:.1%}")
print(f"  Specificity: {specificity:.1%}")
print(f"  False positive rate: {false_positive_rate:.1%}")
```

```python
# Analytical Bayes calculation
# P(Disease | Positive) = P(Pos | Dis) * P(Dis) / P(Pos)
# P(Positive) = P(Pos | Dis) * P(Dis) + P(Pos | No Dis) * P(No Dis)

p_positive = sensitivity * prevalence + false_positive_rate * (1 - prevalence)
p_disease_given_positive = (sensitivity * prevalence) / p_positive

print(f"\nAnalytical (Bayes):")
print(f"  P(Positive) = {p_positive:.4f}")
print(f"  P(Disease | Positive) = {p_disease_given_positive:.4f}")
print(f"  Despite 95% sensitivity, only {p_disease_given_positive:.1%} of positives have disease!")
```

```python
# Simulation verification
n_people = 1000000

# Generate disease status
has_disease = np.random.random(n_people) < prevalence

# Generate test results
test_positive = np.zeros(n_people, dtype=bool)
# If diseased: positive with probability = sensitivity
test_positive[has_disease] = np.random.random(np.sum(has_disease)) < sensitivity
# If not diseased: positive with probability = false_positive_rate
test_positive[~has_disease] = np.random.random(np.sum(~has_disease)) < false_positive_rate

# Count outcomes
true_positives = np.sum(has_disease & test_positive)
false_positives = np.sum(~has_disease & test_positive)
total_positives = true_positives + false_positives

p_disease_given_positive_sim = true_positives / total_positives

print(f"\nSimulation ({n_people:,} people):")
print(f"  True positives: {true_positives:,}")
print(f"  False positives: {false_positives:,}")
print(f"  Total positives: {total_positives:,}")
print(f"  P(Disease | Positive) = {p_disease_given_positive_sim:.4f}")
```

```python
# What if prevalence changes?
print("\nEffect of prevalence on P(Disease | Positive):")
for prev in [0.001, 0.01, 0.05, 0.10, 0.20]:
    p_pos = sensitivity * prev + false_positive_rate * (1 - prev)
    p_dis_given_pos = (sensitivity * prev) / p_pos
    print(f"  Prevalence {prev:5.1%}: P(Disease|Positive) = {p_dis_given_pos:.3f}")
```

```python
# Confusion matrix for the test
true_negatives = np.sum(~has_disease & ~test_positive)
false_negatives = np.sum(has_disease & ~test_positive)

print("\nConfusion Matrix:")
print(f"                  | Positive | Negative |")
print(f"  Has Disease     | {true_positives:8,} | {false_negatives:8,} |")
print(f"  No Disease      | {false_positives:8,} | {true_negatives:8,} |")
```

## Key Observations

- Base rate (prevalence) dramatically affects posterior probability.
- With 1% prevalence, most positive tests are false positives.
- High sensitivity doesn't mean high positive predictive value.
- Simulation confirms Bayes' theorem calculation.
- This is why rare disease screening requires careful interpretation.

# KNN vs SVM

Compare K-Nearest Neighbors and Support Vector Machines.

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
np.random.seed(42)

X, y = make_classification(n_samples=500, n_features=10, n_informative=5, random_state=42)
X = StandardScaler().fit_transform(X)
```

```python
# Compare different K values
for k in [1, 5, 15, 50]:
    knn = KNeighborsClassifier(n_neighbors=k)
    score = cross_val_score(knn, X, y, cv=5).mean()
    print(f"KNN k={k:2d}: {score:.3f}")

print()
for kernel in ['linear', 'rbf', 'poly']:
    svm = SVC(kernel=kernel)
    score = cross_val_score(svm, X, y, cv=5).mean()
    print(f"SVM {kernel}: {score:.3f}")
```

## Key Observations
- KNN: instance-based, needs scaling, k controls smoothness
- SVM: finds max-margin boundary, kernel for non-linearity
- Both need feature scaling

# Objective Functions in ML

Explore common objective functions: MSE, MAE, cross-entropy, and their properties.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Regression losses
errors = np.linspace(-3, 3, 100)

# MSE: penalizes large errors heavily
mse = errors ** 2

# MAE: linear penalty
mae = np.abs(errors)

# Huber: combines MSE and MAE
delta = 1.0
huber = np.where(np.abs(errors) <= delta, 
                 0.5 * errors**2,
                 delta * (np.abs(errors) - 0.5 * delta))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

ax = axes[0]
ax.plot(errors, mse, label='MSE (L2)', linewidth=2)
ax.plot(errors, mae, label='MAE (L1)', linewidth=2)
ax.plot(errors, huber, label=f'Huber (δ={delta})', linewidth=2)
ax.set_xlabel('Error (y - ŷ)')
ax.set_ylabel('Loss')
ax.set_title('Regression Loss Functions')
ax.legend()
ax.grid(True, alpha=0.3)

# Gradients
ax = axes[1]
ax.plot(errors, 2 * errors, label='∂MSE/∂error', linewidth=2)
ax.plot(errors, np.sign(errors), label='∂MAE/∂error', linewidth=2)
ax.set_xlabel('Error')
ax.set_ylabel('Gradient')
ax.set_title('Gradients (for optimization)')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Binary cross-entropy
# L = -[y·log(p) + (1-y)·log(1-p)]

p = np.linspace(0.001, 0.999, 100)

# Loss when true label is 1
bce_y1 = -np.log(p)

# Loss when true label is 0
bce_y0 = -np.log(1 - p)

plt.figure(figsize=(8, 5))
plt.plot(p, bce_y1, label='y=1: -log(p)', linewidth=2)
plt.plot(p, bce_y0, label='y=0: -log(1-p)', linewidth=2)
plt.xlabel('Predicted probability p')
plt.ylabel('Loss')
plt.title('Binary Cross-Entropy')
plt.legend()
plt.grid(True, alpha=0.3)
plt.ylim(0, 5)
plt.show()
```

```python
# Cross-entropy interpretation
print("Binary Cross-Entropy values:")
print("\nWhen true label y = 1:")
for prob in [0.99, 0.9, 0.5, 0.1, 0.01]:
    loss = -np.log(prob)
    print(f"  p = {prob:.2f}: loss = {loss:.3f}")

print("\nWhen true label y = 0:")
for prob in [0.01, 0.1, 0.5, 0.9, 0.99]:
    loss = -np.log(1 - prob)
    print(f"  p = {prob:.2f}: loss = {loss:.3f}")
```

```python
# Effect of outliers: MSE vs MAE
np.random.seed(42)
y_true = np.array([1, 2, 3, 4, 5, 100])  # Last one is outlier
y_pred_good = np.array([1.1, 2.0, 2.9, 4.1, 5.0, 5.0])

errors = y_true - y_pred_good
print(f"Errors: {errors}")
print(f"\nMSE: {np.mean(errors**2):.2f}")
print(f"MAE: {np.mean(np.abs(errors)):.2f}")
print("\nMSE is dominated by the outlier error!")
```

```python
# Multi-class cross-entropy (softmax + CE)
def softmax(logits):
    exp_logits = np.exp(logits - np.max(logits))  # Numerical stability
    return exp_logits / exp_logits.sum()

def cross_entropy(probs, true_class):
    return -np.log(probs[true_class])

logits = np.array([2.0, 1.0, 0.5])  # Raw scores for 3 classes
probs = softmax(logits)

print(f"Logits: {logits}")
print(f"Softmax probabilities: {probs.round(3)}")
print(f"Sum: {probs.sum():.4f}")
print(f"\nCross-entropy if true class is:")
for i in range(3):
    print(f"  Class {i}: {cross_entropy(probs, i):.3f}")
```

## Key Observations

- MSE penalizes large errors quadratically (sensitive to outliers).
- MAE has constant gradient (robust to outliers).
- Cross-entropy heavily penalizes confident wrong predictions.
- Softmax converts logits to probabilities summing to 1.
- Choice of loss function should match the problem and data characteristics.

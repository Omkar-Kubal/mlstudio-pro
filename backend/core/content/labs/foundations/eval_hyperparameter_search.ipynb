# Hyperparameter Search

Compare Grid Search and Random Search for tuning.

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from scipy.stats import uniform, loguniform
np.random.seed(42)

X, y = make_classification(n_samples=200, random_state=42)
```

```python
# Grid Search
param_grid = {'C': [0.1, 1, 10], 'gamma': [0.01, 0.1, 1]}
grid = GridSearchCV(SVC(), param_grid, cv=3, scoring='accuracy')
grid.fit(X, y)
print(f"Grid Search: {grid.best_params_}, score={grid.best_score_:.3f}")
print(f"Grid evaluations: {len(grid.cv_results_['mean_test_score'])}")

# Random Search
param_dist = {'C': loguniform(0.01, 100), 'gamma': loguniform(0.001, 10)}
random = RandomizedSearchCV(SVC(), param_dist, n_iter=10, cv=3, random_state=42)
random.fit(X, y)
print(f"\nRandom Search: C={random.best_params_['C']:.3f}, gamma={random.best_params_['gamma']:.4f}")
print(f"Random best score: {random.best_score_:.3f}")
```

## Key Observations
- Grid: exhaustive, exponential in parameters
- Random: samples from distributions, more efficient
- Random often finds better values in fewer iterations
- Use loguniform for hyperparams spanning orders of magnitude

# Data Leakage Examples

Identify and avoid common data leakage scenarios.

```python
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
np.random.seed(42)

X = np.random.randn(1000, 5)
y = (X[:, 0] > 0).astype(int)
```

```python
# WRONG: Scale before splitting (leakage!)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)  # Uses ALL data stats
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2)
model = LogisticRegression().fit(X_train, y_train)
print(f"Leaky accuracy: {model.score(X_test, y_test):.4f}")

# CORRECT: Split first, then scale
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)  # Fit only on train
X_test_scaled = scaler.transform(X_test)  # Transform test
model = LogisticRegression().fit(X_train_scaled, y_train)
print(f"Proper accuracy: {model.score(X_test_scaled, y_test):.4f}")
```

## Key Observations
- Leakage: test data information bleeds into training
- Common causes: scaling before split, future information
- Always split first, then preprocess

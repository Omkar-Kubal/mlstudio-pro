# Variance and Standard Deviation

Understand how variance and standard deviation quantify spread. Compare datasets with same mean but different spreads.

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
```

```python
# Three datasets with same mean, different spread
mean = 50
n = 500

low_spread = np.random.normal(mean, 5, n)
med_spread = np.random.normal(mean, 15, n)
high_spread = np.random.normal(mean, 30, n)

print(f"Low spread:  Mean={np.mean(low_spread):.2f}, Std={np.std(low_spread):.2f}")
print(f"Med spread:  Mean={np.mean(med_spread):.2f}, Std={np.std(med_spread):.2f}")
print(f"High spread: Mean={np.mean(high_spread):.2f}, Std={np.std(high_spread):.2f}")
```

```python
# Manual variance calculation
data = np.array([2, 4, 4, 4, 5, 5, 7, 9])
n = len(data)
mean = np.mean(data)

# Step by step
deviations = data - mean
squared_deviations = deviations ** 2
variance_pop = np.mean(squared_deviations)  # Population variance
variance_sample = np.sum(squared_deviations) / (n - 1)  # Sample variance (Bessel)

print(f"\nManual calculation on {data}:")
print(f"  Mean: {mean}")
print(f"  Deviations: {deviations}")
print(f"  Squared deviations: {squared_deviations}")
print(f"  Population variance: {variance_pop:.2f}")
print(f"  Sample variance (n-1): {variance_sample:.2f}")
print(f"  Std (population): {np.sqrt(variance_pop):.2f}")
```

```python
# Verify with NumPy
print(f"\nNumPy verification:")
print(f"  np.var(data, ddof=0): {np.var(data, ddof=0):.2f}")
print(f"  np.var(data, ddof=1): {np.var(data, ddof=1):.2f}")
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

bins = np.linspace(-50, 150, 50)
ax.hist(low_spread, bins=bins, alpha=0.5, label=f'σ≈5', edgecolor='black')
ax.hist(med_spread, bins=bins, alpha=0.5, label=f'σ≈15', edgecolor='black')
ax.hist(high_spread, bins=bins, alpha=0.5, label=f'σ≈30', edgecolor='black')
ax.axvline(mean, color='black', linestyle='--', label='Mean=50')
ax.set_xlabel('Value')
ax.set_ylabel('Frequency')
ax.set_title('Same Mean, Different Spread')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Empirical rule verification (68-95-99.7)
data = np.random.normal(100, 15, 10000)
mean, std = np.mean(data), np.std(data)

within_1_std = np.sum((data >= mean - std) & (data <= mean + std)) / len(data)
within_2_std = np.sum((data >= mean - 2*std) & (data <= mean + 2*std)) / len(data)
within_3_std = np.sum((data >= mean - 3*std) & (data <= mean + 3*std)) / len(data)

print(f"\nEmpirical Rule (Normal data):")
print(f"  Within 1 σ: {within_1_std:.1%} (expect ~68%)")
print(f"  Within 2 σ: {within_2_std:.1%} (expect ~95%)")
print(f"  Within 3 σ: {within_3_std:.1%} (expect ~99.7%)")
```

## Key Observations

- Variance = average of squared deviations from mean.
- Standard deviation = √variance, in original units.
- Same mean, different std means very different distributions.
- Sample variance uses n-1 (Bessel's correction) for unbiased estimation.
- For normal data, ~68% within 1σ, ~95% within 2σ, ~99.7% within 3σ.

# Weight Initialization Effects

Show how initialization affects training.

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)

def sigmoid(x): return 1 / (1 + np.exp(-np.clip(x, -500, 500)))

def forward_pass(X, weights):
    a = X
    activations = [a]
    for W in weights:
        z = a @ W
        a = sigmoid(z)
        activations.append(a)
    return activations
```

```python
n_layers = 10
n_neurons = 100

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

for ax, scale, title in [(axes[0], 0.01, 'Too Small (0.01)'), (axes[1], 1.0, 'Too Large (1.0)'), (axes[2], np.sqrt(2/n_neurons), 'Xavier')]:
    weights = [np.random.randn(n_neurons, n_neurons) * scale for _ in range(n_layers)]
    X = np.random.randn(1000, n_neurons)
    acts = forward_pass(X, weights)
    
    for i, a in enumerate(acts):
        ax.hist(a.ravel(), bins=50, alpha=0.5, label=f'Layer {i}')
    ax.set_title(f'{title}, std={scale:.4f}')
    ax.set_xlabel('Activation value')

plt.tight_layout()
plt.show()
```

## Key Observations
- Too small: activations vanish to 0
- Too large: activations saturate to 0 or 1
- Xavier/He: maintains activation variance
- Bad init = vanishing/exploding gradients

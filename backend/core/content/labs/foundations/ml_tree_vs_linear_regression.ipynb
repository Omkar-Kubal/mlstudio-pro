# Tree vs Linear Regression

Compare linear models and decision trees on non-linear data.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
np.random.seed(42)

X = np.linspace(0, 10, 100).reshape(-1, 1)
y = np.sin(X.ravel()) + np.random.randn(100)*0.2
```

```python
lr = LinearRegression().fit(X, y)
dt = DecisionTreeRegressor(max_depth=4).fit(X, y)

plt.figure(figsize=(10, 5))
plt.scatter(X, y, alpha=0.5, label='Data')
plt.plot(X, lr.predict(X), 'r-', label='Linear')
plt.plot(X, dt.predict(X), 'g-', label='Tree (depth=4)')
plt.legend()
plt.title('Linear vs Tree on Non-linear Data')
plt.show()

print(f"Linear R²: {lr.score(X, y):.3f}")
print(f"Tree R²: {dt.score(X, y):.3f}")
```

## Key Observations
- Linear: single line, poor for non-linear patterns
- Tree: step function, captures non-linearity
- Tree depth controls complexity

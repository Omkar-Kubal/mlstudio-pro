# Descriptive vs Inferential Statistics

Demonstrate how descriptive statistics summarize a sample, while inferential statistics estimate population parameters with uncertainty.

```python
import numpy as np
np.random.seed(42)
```

```python
# True population (unknown in practice)
population_mean = 100
population_std = 15
population = np.random.normal(population_mean, population_std, size=100000)

print(f"True population mean: {population_mean}")
print(f"True population std: {population_std}")
```

```python
# Draw a sample
sample_size = 50
sample = np.random.choice(population, size=sample_size, replace=False)

# Descriptive: summarize the sample
sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)  # Bessel's correction

print(f"\nDESCRIPTIVE (sample of n={sample_size}):")
print(f"  Sample mean: {sample_mean:.2f}")
print(f"  Sample std: {sample_std:.2f}")
```

```python
# Inferential: estimate population mean with confidence interval
from scipy import stats

standard_error = sample_std / np.sqrt(sample_size)
confidence_level = 0.95
t_critical = stats.t.ppf((1 + confidence_level) / 2, df=sample_size - 1)
margin_of_error = t_critical * standard_error

ci_lower = sample_mean - margin_of_error
ci_upper = sample_mean + margin_of_error

print(f"\nINFERENTIAL (estimate population):")
print(f"  Point estimate of population mean: {sample_mean:.2f}")
print(f"  95% CI: [{ci_lower:.2f}, {ci_upper:.2f}]")
print(f"  True mean captured: {ci_lower <= population_mean <= ci_upper}")
```

```python
# Repeat sampling to show CI coverage
n_simulations = 1000
captured = 0

for _ in range(n_simulations):
    s = np.random.choice(population, size=sample_size, replace=False)
    s_mean = np.mean(s)
    s_std = np.std(s, ddof=1)
    se = s_std / np.sqrt(sample_size)
    ci_l = s_mean - t_critical * se
    ci_u = s_mean + t_critical * se
    if ci_l <= population_mean <= ci_u:
        captured += 1

print(f"\nCI coverage over {n_simulations} samples: {captured / n_simulations:.1%}")
```

## Key Observations

- Descriptive statistics describe what you have (the sample).
- Inferential statistics generalize to what you don't have (the population).
- Confidence intervals quantify uncertainty in estimates.
- ~95% of 95% CIs capture the true parameter (by design).

# Probability Simulation

Simulate basic probability experiments to verify theoretical probabilities through repeated trials.

```python
import numpy as np
np.random.seed(42)
```

```python
# Coin flip: P(heads) = 0.5
n_flips = 10000
flips = np.random.choice(['H', 'T'], size=n_flips)
p_heads = np.mean(flips == 'H')
print(f"Coin flips ({n_flips} trials): P(heads) = {p_heads:.4f} (expected 0.5)")
```

```python
# Dice roll: P(6) = 1/6 ≈ 0.1667
n_rolls = 10000
rolls = np.random.randint(1, 7, size=n_rolls)
p_six = np.mean(rolls == 6)
print(f"Dice rolls ({n_rolls} trials): P(6) = {p_six:.4f} (expected 0.1667)")
```

```python
# Sum of two dice: P(sum=7)
n_trials = 10000
die1 = np.random.randint(1, 7, n_trials)
die2 = np.random.randint(1, 7, n_trials)
sums = die1 + die2

p_sum_7 = np.mean(sums == 7)
print(f"Two dice sum ({n_trials} trials): P(sum=7) = {p_sum_7:.4f} (expected 6/36 = 0.1667)")

# Distribution of sums
for s in range(2, 13):
    print(f"  P(sum={s:2d}) = {np.mean(sums == s):.4f}")
```

```python
# Birthday problem: P(at least 2 share birthday in group of n)
def simulate_birthday(n_people, n_simulations=10000):
    matches = 0
    for _ in range(n_simulations):
        birthdays = np.random.randint(0, 365, n_people)
        if len(birthdays) != len(set(birthdays)):
            matches += 1
    return matches / n_simulations

print("\nBirthday Problem:")
for n in [10, 23, 30, 50, 70]:
    p = simulate_birthday(n)
    print(f"  n={n:2d} people: P(match) = {p:.3f}")
```

```python
# Monty Hall problem simulation
def simulate_monty_hall(switch, n_games=10000):
    wins = 0
    for _ in range(n_games):
        # Car behind one of 3 doors
        car = np.random.randint(0, 3)
        # Player picks a door
        pick = np.random.randint(0, 3)
        # Host opens a door with goat (not car, not picked)
        available = [d for d in range(3) if d != pick and d != car]
        # If switch, player switches to remaining door
        if switch:
            remaining = [d for d in range(3) if d != pick and d not in available[:1]][0]
            pick = remaining
        if pick == car:
            wins += 1
    return wins / n_games

print("\nMonty Hall Problem:")
print(f"  Stay strategy:   P(win) = {simulate_monty_hall(switch=False):.3f} (expected 1/3)")
print(f"  Switch strategy: P(win) = {simulate_monty_hall(switch=True):.3f} (expected 2/3)")
```

## Key Observations

- Simulated frequencies converge to theoretical probabilities with many trials.
- Birthday problem: only 23 people needed for >50% probability of shared birthday.
- Monty Hall: switching doubles your chances (counterintuitive but correct).
- Simulation is a powerful tool for verifying complex probability calculations.

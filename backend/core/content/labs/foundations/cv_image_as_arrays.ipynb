# Image as Arrays

Understand image representation as numerical arrays.

```python
import numpy as np
import matplotlib.pyplot as plt

# Grayscale: 2D array (H, W)
gray = np.zeros((100, 100))
gray[20:80, 20:80] = 255  # White square

# RGB: 3D array (H, W, 3)
rgb = np.zeros((100, 100, 3), dtype=np.uint8)
rgb[20:50, 20:50] = [255, 0, 0]   # Red
rgb[50:80, 50:80] = [0, 0, 255]   # Blue
rgb[20:50, 50:80] = [0, 255, 0]   # Green
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].imshow(gray, cmap='gray')
axes[0].set_title(f'Grayscale {gray.shape}')
axes[1].imshow(rgb)
axes[1].set_title(f'RGB {rgb.shape}')
plt.show()

print(f"Pixel value at (25,25): Gray={gray[25,25]}, RGB={rgb[25,25]}")
```

## Key Observations
- Grayscale: (H, W) with values 0-255
- RGB: (H, W, 3) for Red, Green, Blue channels
- CNNs process images as multi-channel tensors
- Normalize to [0,1] or [-1,1] for training

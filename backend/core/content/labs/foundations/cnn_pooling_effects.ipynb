# Pooling Effects

Compare max pooling and average pooling.

```python
import numpy as np
import matplotlib.pyplot as plt

def max_pool(x, size=2):
    h, w = x.shape
    out = np.zeros((h//size, w//size))
    for i in range(out.shape[0]):
        for j in range(out.shape[1]):
            out[i,j] = np.max(x[i*size:(i+1)*size, j*size:(j+1)*size])
    return out

np.random.seed(42)
fm = np.random.rand(8, 8)
print(f"Original: {fm.shape} → Pooled: {max_pool(fm).shape}")
```

## Key Observations
- Reduces spatial dimensions by factor of pool size
- Max pool preserves strongest activations
- Introduces translation invariance

# Visualizing Model Results

Create plots for model evaluation: learning curves, predictions vs actuals, feature importance.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import learning_curve
np.random.seed(42)
```

```python
# Generate data and train model
X = np.random.randn(100, 3)
y = 2*X[:, 0] + 3*X[:, 1] - X[:, 2] + np.random.randn(100)*0.5

model = LinearRegression().fit(X, y)
y_pred = model.predict(X)
```

```python
# Predictions vs Actuals
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

ax = axes[0]
ax.scatter(y, y_pred, alpha=0.7)
ax.plot([y.min(), y.max()], [y.min(), y.max()], 'r--')
ax.set_xlabel('Actual')
ax.set_ylabel('Predicted')
ax.set_title('Predictions vs Actuals')

# Residuals
ax = axes[1]
residuals = y - y_pred
ax.scatter(y_pred, residuals, alpha=0.7)
ax.axhline(0, color='red', linestyle='--')
ax.set_xlabel('Predicted')
ax.set_ylabel('Residual')
ax.set_title('Residual Plot')

plt.tight_layout()
plt.show()
```

```python
# Feature importance
importance = np.abs(model.coef_)
features = ['Feature 1', 'Feature 2', 'Feature 3']

plt.figure(figsize=(8, 4))
plt.barh(features, importance)
plt.xlabel('Absolute Coefficient')
plt.title('Feature Importance')
plt.show()
```

```python
# Learning curve
train_sizes, train_scores, val_scores = learning_curve(
    LinearRegression(), X, y, train_sizes=np.linspace(0.1, 1.0, 10), cv=5)

plt.figure(figsize=(8, 5))
plt.plot(train_sizes, train_scores.mean(axis=1), label='Train')
plt.plot(train_sizes, val_scores.mean(axis=1), label='Validation')
plt.xlabel('Training Size')
plt.ylabel('Score (R²)')
plt.title('Learning Curve')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Key Observations
- Prediction vs actual: points on diagonal = good fit
- Residual plot: random scatter = good; patterns = problems
- Learning curves: diagnose bias (both low) vs variance (gap)
- Feature importance: which features drive predictions

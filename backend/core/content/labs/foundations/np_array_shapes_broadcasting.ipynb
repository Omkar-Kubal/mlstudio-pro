# Array Shapes and Broadcasting

Understand NumPy array shapes and broadcasting rules for efficient computation.

```python
import numpy as np
```

```python
# Array shapes
scalar = np.array(5)           # shape: ()
vector = np.array([1, 2, 3])   # shape: (3,)
matrix = np.array([[1, 2], [3, 4], [5, 6]])  # shape: (3, 2)
tensor = np.zeros((2, 3, 4))   # shape: (2, 3, 4)

print(f"Scalar shape: {scalar.shape}, ndim: {scalar.ndim}")
print(f"Vector shape: {vector.shape}, ndim: {vector.ndim}")
print(f"Matrix shape: {matrix.shape}, ndim: {matrix.ndim}")
print(f"Tensor shape: {tensor.shape}, ndim: {tensor.ndim}")
```

```python
# Reshaping
a = np.arange(12)
print(f"Original: {a.shape}")
print(f"Reshape (3,4): {a.reshape(3, 4).shape}")
print(f"Reshape (2,2,3): {a.reshape(2, 2, 3).shape}")
print(f"Reshape (-1,4): {a.reshape(-1, 4).shape}")
```

```python
# Broadcasting: scalar + array
arr = np.array([1, 2, 3])
print(f"{arr} + 10 = {arr + 10}")

# Broadcasting: (3,1) + (1,4) = (3,4)
col = np.array([[1], [2], [3]])  # (3, 1)
row = np.array([[10, 20, 30, 40]])  # (1, 4)
result = col + row  # (3, 4)

print(f"\nCol (3,1):\n{col}")
print(f"Row (1,4): {row}")
print(f"Result (3,4):\n{result}")
```

```python
# Broadcasting rules
# 1. Align shapes from right
# 2. Dimensions must be equal OR one of them is 1

a = np.ones((4, 3))
b = np.ones((3,))
c = np.ones((4, 1))

print(f"(4,3) + (3,) = {(a + b).shape}")    # Works
print(f"(4,3) + (4,1) = {(a + c).shape}")  # Works

# This would fail:
# d = np.ones((4,))
# a + d  # (4,3) + (4,) fails - 3 ≠ 4
```

```python
# Practical example: normalize columns
data = np.array([[1, 100], [2, 200], [3, 300]])
col_means = data.mean(axis=0)  # (2,)
col_stds = data.std(axis=0)    # (2,)

normalized = (data - col_means) / col_stds  # Broadcasting!
print(f"Normalized:\n{normalized.round(2)}")
```

## Key Observations
- shape = dimensions (rows, cols, depth, ...)
- reshape(-1, n) infers dimension automatically
- Broadcasting: align from right, dim must be 1 or equal
- Broadcasting enables vectorized operations without loops

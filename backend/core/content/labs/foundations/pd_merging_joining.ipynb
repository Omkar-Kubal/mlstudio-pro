# Merging and Joining DataFrames

Combine DataFrames using merge, join, and concat operations.

```python
import pandas as pd

# Sample data
employees = pd.DataFrame({
    'emp_id': [1, 2, 3, 4],
    'name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'dept_id': [10, 20, 10, 30]
})

departments = pd.DataFrame({
    'dept_id': [10, 20, 40],
    'dept_name': ['Sales', 'IT', 'Marketing']
})

print("Employees:\n", employees)
print("\nDepartments:\n", departments)
```

```python
# Inner join (only matching)
inner = pd.merge(employees, departments, on='dept_id', how='inner')
print(f"Inner join:\n{inner}")
```

```python
# Left join (all from left)
left = pd.merge(employees, departments, on='dept_id', how='left')
print(f"Left join:\n{left}")
```

```python
# Outer join (all from both)
outer = pd.merge(employees, departments, on='dept_id', how='outer')
print(f"Outer join:\n{outer}")
```

```python
# Concat (stacking DataFrames)
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})

vertical = pd.concat([df1, df2], ignore_index=True)
horizontal = pd.concat([df1, df2], axis=1)

print(f"Vertical concat:\n{vertical}")
print(f"\nHorizontal concat:\n{horizontal}")
```

## Key Observations
- inner: only matching keys
- left: all from left, match or NaN from right
- outer: all from both, NaN where no match
- concat for stacking, merge for database-style joins

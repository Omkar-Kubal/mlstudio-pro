# Skewness and Kurtosis

Compute and visualize skewness (asymmetry) and kurtosis (tail heaviness) to understand distribution shapes beyond mean and variance.

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
np.random.seed(42)
```

```python
# Generate distributions with different skewness
n = 5000
symmetric = np.random.normal(50, 10, n)
right_skew = np.random.exponential(10, n)
left_skew = 100 - np.random.exponential(10, n)

print("SKEWNESS (0 = symmetric):")
print(f"  Symmetric:    {stats.skew(symmetric):.3f}")
print(f"  Right-skewed: {stats.skew(right_skew):.3f}")
print(f"  Left-skewed:  {stats.skew(left_skew):.3f}")
```

```python
# Generate distributions with different kurtosis
normal = np.random.normal(0, 1, n)  # kurtosis ≈ 0 (excess)
uniform = np.random.uniform(-2, 2, n)  # platykurtic (light tails)
heavy_tails = np.random.standard_t(df=3, size=n)  # leptokurtic (heavy tails)

print("\nKURTOSIS (excess, 0 = normal):")
print(f"  Normal (mesokurtic):  {stats.kurtosis(normal):.3f}")
print(f"  Uniform (platykurtic): {stats.kurtosis(uniform):.3f}")
print(f"  t(df=3) (leptokurtic): {stats.kurtosis(heavy_tails):.3f}")
```

```python
fig, axes = plt.subplots(2, 3, figsize=(14, 8))

# Skewness row
for ax, data, title in zip(axes[0], 
                           [symmetric, right_skew, left_skew],
                           ['Symmetric (skew≈0)', 'Right-skewed (skew>0)', 'Left-skewed (skew<0)']):
    ax.hist(data, bins=40, alpha=0.7, edgecolor='black')
    ax.axvline(np.mean(data), color='red', linestyle='--', label='Mean')
    ax.axvline(np.median(data), color='green', linestyle='-', label='Median')
    ax.set_title(title)
    ax.legend(fontsize=8)

# Kurtosis row
for ax, data, title in zip(axes[1],
                           [uniform, normal, heavy_tails],
                           ['Platykurtic (light tails)', 'Mesokurtic (normal)', 'Leptokurtic (heavy tails)']):
    ax.hist(data, bins=50, alpha=0.7, edgecolor='black', range=(-6, 6))
    ax.set_title(title)
    ax.set_xlim(-6, 6)

plt.tight_layout()
plt.show()
```

```python
# Real-world example: income distribution
income = np.random.lognormal(mean=10.5, sigma=0.8, size=10000)
print(f"\nIncome distribution:")
print(f"  Skewness: {stats.skew(income):.2f} (highly right-skewed)")
print(f"  Kurtosis: {stats.kurtosis(income):.2f} (heavy right tail)")
```

## Key Observations

- Skewness measures asymmetry: positive = right tail longer, negative = left tail longer.
- Kurtosis measures tail heaviness: positive = heavier than normal, negative = lighter.
- Right-skewed: mean > median. Left-skewed: mean < median.
- High kurtosis means more outliers (extreme values more likely).
- These metrics help identify non-normal data before applying statistical tests.

# Lab: Transformer Forward Pass (Core)

Implementation of the mathematical forward pass of a Transformer block using NumPy.

```python
import numpy as np
np.random.seed(42)
def attention(q, k, v):
    d_k = q.shape[-1]
    scores = np.matmul(q, k.swapaxes(-2, -1)) / np.sqrt(d_k)
    weights = np.exp(scores - np.max(scores, axis=-1, keepdims=True))
    weights = weights / weights.sum(axis=-1, keepdims=True)
    return np.matmul(weights, v), weights
q = k = v = np.random.randn(3, 4)
out, w = attention(q, k, v)
print("Done. Hashable output:", np.round(out.sum(), 4))
```

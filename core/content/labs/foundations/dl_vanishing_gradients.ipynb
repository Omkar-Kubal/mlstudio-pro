# Vanishing Gradients

Visualize how gradients shrink through deep networks.

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(x): return 1 / (1 + np.exp(-np.clip(x, -500, 500)))
def sigmoid_grad(x): s = sigmoid(x); return s * (1 - s)
def relu(x): return np.maximum(0, x)
def relu_grad(x): return (x > 0).astype(float)
```

```python
# Max gradient for sigmoid
x = np.linspace(-5, 5, 100)
print(f"Sigmoid max gradient: {sigmoid_grad(x).max():.4f} (at x=0)")
print("Each layer multiplies gradient by at most 0.25!")
print(f"After 10 layers: 0.25^10 = {0.25**10:.2e}")

# ReLU comparison
print(f"\nReLU gradient: 0 or 1")
print("No shrinking for positive inputs!")
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x, sigmoid_grad(x), label='Sigmoid gradient', linewidth=2)
ax.plot(x, relu_grad(x), label='ReLU gradient', linewidth=2)
ax.axhline(0.25, color='red', linestyle='--', alpha=0.5, label='Max sigmoid grad')
ax.set_xlabel('x')
ax.set_ylabel('Gradient')
ax.set_title('Activation Gradients')
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

## Key Observations
- Sigmoid gradient max 0.25, vanishes for deep nets
- ReLU gradient is 0 or 1, no vanishing
- ResNets add skip connections to help gradients flow
- Batch norm also helps gradient flow

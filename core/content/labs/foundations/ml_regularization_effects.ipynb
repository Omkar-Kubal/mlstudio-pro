# Regularization Effects

Visualize how Ridge and Lasso shrink coefficients.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Ridge, Lasso, LinearRegression
np.random.seed(42)

X = np.random.randn(100, 10)
true_w = np.array([3, 2, 1, 0, 0, 0, 0, 0, 0, 0])
y = X @ true_w + np.random.randn(100)*0.5
```

```python
alphas = [0.01, 0.1, 1, 10, 100]
ridge_coefs = []
lasso_coefs = []

for a in alphas:
    ridge_coefs.append(Ridge(alpha=a).fit(X, y).coef_)
    lasso_coefs.append(Lasso(alpha=a/10).fit(X, y).coef_)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for coefs, ax, title in [(ridge_coefs, axes[0], 'Ridge'), (lasso_coefs, axes[1], 'Lasso')]:
    for i, c in enumerate(coefs):
        ax.bar(np.arange(10) + i*0.15, c, width=0.15, label=f'α={alphas[i]}')
    ax.axhline(0, color='black', linewidth=0.5)
    ax.set_title(title)
    ax.legend(fontsize=8)

plt.tight_layout()
plt.show()
```

## Key Observations
- Ridge: shrinks all coefficients toward 0
- Lasso: drives some coefficients exactly to 0 (feature selection)
- Higher α = more regularization

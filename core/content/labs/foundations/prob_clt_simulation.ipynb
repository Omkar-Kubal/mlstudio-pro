# Central Limit Theorem Simulation

Demonstrate that sample means from any distribution converge to a normal distribution as sample size increases.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
np.random.seed(42)
```

```python
def simulate_clt(population_generator, pop_name, sample_sizes, n_samples=5000):
    """Simulate CLT for a given population distribution."""
    fig, axes = plt.subplots(1, len(sample_sizes), figsize=(14, 4))
    
    # Get population stats
    pop = population_generator(100000)
    pop_mean, pop_std = np.mean(pop), np.std(pop)
    
    for ax, n in zip(axes, sample_sizes):
        # Draw many samples and compute means
        sample_means = [np.mean(population_generator(n)) for _ in range(n_samples)]
        
        # Plot histogram
        ax.hist(sample_means, bins=40, density=True, alpha=0.7, edgecolor='black')
        
        # Overlay theoretical normal
        expected_se = pop_std / np.sqrt(n)
        x = np.linspace(pop_mean - 4*expected_se, pop_mean + 4*expected_se, 100)
        ax.plot(x, stats.norm.pdf(x, pop_mean, expected_se), 'r-', lw=2)
        
        ax.set_title(f'n = {n}')
        ax.set_xlabel('Sample Mean')
    
    fig.suptitle(f'CLT: {pop_name}', fontsize=12, y=1.02)
    plt.tight_layout()
    plt.show()
```

```python
# Uniform distribution (not normal, flat)
uniform_gen = lambda n: np.random.uniform(0, 1, n)
simulate_clt(uniform_gen, 'Uniform(0,1)', [1, 5, 30, 100])
```

```python
# Exponential distribution (highly skewed)
exp_gen = lambda n: np.random.exponential(1, n)
simulate_clt(exp_gen, 'Exponential(λ=1)', [1, 5, 30, 100])
```

```python
# Bernoulli (discrete, binary)
bernoulli_gen = lambda n: np.random.binomial(1, 0.3, n)
simulate_clt(bernoulli_gen, 'Bernoulli(p=0.3)', [1, 5, 30, 100])
```

```python
# Verify normality with statistical test
n = 30
n_samples = 1000

# Sample means from exponential
means = [np.mean(np.random.exponential(1, n)) for _ in range(n_samples)]

# Shapiro-Wilk test for normality
stat, p_value = stats.shapiro(means[:500])  # Test works best with smaller samples
print(f"Shapiro-Wilk test on sample means (n={n}):")
print(f"  Statistic = {stat:.4f}")
print(f"  p-value = {p_value:.4f}")
print(f"  Normal? {'Yes' if p_value > 0.05 else 'No'} (at α=0.05)")
```

```python
# Standard error formula verification
pop = np.random.exponential(10, 100000)
pop_std = np.std(pop)

print("\nSE verification (Exponential population):")
print(f"Population std: {pop_std:.3f}")
for n in [10, 25, 50, 100, 400]:
    means = [np.mean(np.random.choice(pop, n)) for _ in range(1000)]
    observed_se = np.std(means)
    theoretical_se = pop_std / np.sqrt(n)
    print(f"  n={n:3d}: Observed SE={observed_se:.3f}, Theoretical={theoretical_se:.3f}")
```

## Key Observations

- Sample means become normally distributed regardless of population shape.
- Convergence is faster for symmetric populations.
- n ≥ 30 is often sufficient for practical normality.
- Mean of sampling distribution = population mean.
- SE = σ/√n quantifies spread of sample means.

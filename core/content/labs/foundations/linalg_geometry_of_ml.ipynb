# Geometry of Machine Learning

Connect linear algebra concepts to ML: feature spaces, projections, and the geometry of linear regression.

```python
import numpy as np
import matplotlib.pyplot as plt
np.set_printoptions(precision=4, suppress=True)
```

```python
# Data points as vectors in feature space
# Each row is a data point, each column is a feature
X = np.array([[1, 2],
              [2, 3],
              [3, 5],
              [4, 4],
              [5, 6]])

print(f"Data matrix X:\n{X}")
print(f"\nShape: {X.shape} (5 samples, 2 features)")
```

```python
# Centering data (important for PCA, covariance)
X_centered = X - X.mean(axis=0)
print(f"Centered X (mean = 0):\n{X_centered}")
print(f"\nColumn means: {X_centered.mean(axis=0)}")
```

```python
# Covariance matrix from centered data
# Cov = X_centered.T @ X_centered / (n-1)
n = X.shape[0]
cov_manual = X_centered.T @ X_centered / (n - 1)
cov_numpy = np.cov(X.T)

print(f"Covariance matrix (manual):\n{cov_manual}")
print(f"\nCovariance matrix (numpy):\n{cov_numpy}")
```

```python
# Linear regression as projection
# y = Xw, find w that minimizes ||y - Xw||²
# Solution: w = (X.T @ X)^(-1) @ X.T @ y

np.random.seed(42)
X_reg = np.column_stack([np.ones(10), np.arange(10)])  # Add bias column
y = 2 + 3 * np.arange(10) + np.random.randn(10)  # y = 2 + 3x + noise

# Normal equations
w = np.linalg.inv(X_reg.T @ X_reg) @ X_reg.T @ y

print(f"Linear regression coefficients: {w}")
print(f"(Expected: intercept ≈ 2, slope ≈ 3)")
```

```python
# Prediction = projection onto column space of X
y_pred = X_reg @ w
residuals = y - y_pred

# Residuals are orthogonal to column space
print(f"X.T @ residuals (should be ≈ 0):")
print(X_reg.T @ residuals)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Regression fit
ax = axes[0]
ax.scatter(X_reg[:, 1], y, label='Data')
ax.plot(X_reg[:, 1], y_pred, 'r-', label=f'Fit: y = {w[0]:.2f} + {w[1]:.2f}x')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
ax.set_title('Linear Regression')

# Residuals
ax = axes[1]
ax.scatter(X_reg[:, 1], residuals)
ax.axhline(0, color='red', linestyle='--')
ax.set_xlabel('x')
ax.set_ylabel('Residual (y - ŷ)')
ax.set_title('Residuals (orthogonal to X)')

plt.tight_layout()
plt.show()
```

```python
# Distance in feature space
p1 = np.array([1, 2, 3])
p2 = np.array([4, 5, 6])

euclidean = np.linalg.norm(p1 - p2)
manhattan = np.sum(np.abs(p1 - p2))

print(f"\nDistances between {p1} and {p2}:")
print(f"  Euclidean (L2): {euclidean:.4f}")
print(f"  Manhattan (L1): {manhattan}")
```

```python
# Hyperplane separation (basis of SVM)
# w · x + b = 0 defines a hyperplane
w_plane = np.array([1, 1])  # Normal vector
b = -3  # Bias

# Points
points = np.array([[1, 1], [2, 2], [3, 3], [1, 3], [3, 1]])
distances_to_plane = (points @ w_plane + b) / np.linalg.norm(w_plane)

print(f"\nSigned distances to hyperplane w·x + b = 0:")
for i, (p, d) in enumerate(zip(points, distances_to_plane)):
    side = 'above' if d > 0 else 'below'
    print(f"  Point {p}: distance = {d:.3f} ({side})")
```

## Key Observations

- Data points live in a feature space (n features = n dimensions).
- Linear regression projects y onto the column space of X.
- Residuals are orthogonal to features (normal equations).
- Covariance matrix captures feature relationships.
- Hyperplanes separate classes in classification (SVM, logistic regression).

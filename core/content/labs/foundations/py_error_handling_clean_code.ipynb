# Error Handling and Clean Code

Write robust code with proper error handling and clean coding practices.

```python
# Try-except basics
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return float('inf')
    except TypeError as e:
        print(f"Type error: {e}")
        return None

print(f"10/2 = {safe_divide(10, 2)}")
print(f"10/0 = {safe_divide(10, 0)}")
print(f"10/'a' = {safe_divide(10, 'a')}")
```

```python
# Context managers (with statement)
# File handling
try:
    with open('test.txt', 'w') as f:
        f.write('Hello World')
    with open('test.txt', 'r') as f:
        print(f.read())
except IOError as e:
    print(f"IO Error: {e}")
```

```python
# Assertions for debugging
def train_model(X, y, lr):
    assert len(X) == len(y), "X and y must have same length"
    assert lr > 0, "Learning rate must be positive"
    print(f"Training with {len(X)} samples, lr={lr}")

train_model([1,2,3], [1,2,3], 0.01)

try:
    train_model([1,2,3], [1,2], 0.01)  # Will fail
except AssertionError as e:
    print(f"AssertionError: {e}")
```

```python
# Type hints (Python 3.5+)
from typing import List, Optional

def compute_mean(values: List[float]) -> Optional[float]:
    """Compute mean of values, return None if empty."""
    if not values:
        return None
    return sum(values) / len(values)

print(compute_mean([1, 2, 3]))
print(compute_mean([]))
```

```python
# Clean code principles
# BAD
def p(d):
    return [x*2 for x in d if x>0]

# GOOD
def filter_and_double_positives(data: List[float]) -> List[float]:
    """Filter positive values and double them."""
    positive_values = [x for x in data if x > 0]
    doubled_values = [x * 2 for x in positive_values]
    return doubled_values

print(filter_and_double_positives([-1, 2, -3, 4]))
```

## Key Observations
- Use try-except for expected errors
- Use assertions for debugging invariants
- Use context managers (with) for resources
- Type hints improve readability and IDE support
- Clear names > comments > clever code

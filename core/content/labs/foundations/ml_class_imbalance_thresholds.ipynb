# Class Imbalance and Thresholds

Handle imbalanced datasets and tune classification thresholds.

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, precision_recall_curve
import matplotlib.pyplot as plt
np.random.seed(42)

X, y = make_classification(n_samples=1000, weights=[0.95, 0.05], random_state=42)
print(f"Class distribution: {np.bincount(y)}")
```

```python
model = LogisticRegression(class_weight='balanced').fit(X, y)
y_prob = model.predict_proba(X)[:, 1]

# Default threshold 0.5
y_pred_50 = (y_prob >= 0.5).astype(int)
# Lower threshold for higher recall
y_pred_30 = (y_prob >= 0.3).astype(int)

print("Threshold 0.5:")
print(classification_report(y, y_pred_50, target_names=['Neg', 'Pos']))
print("\nThreshold 0.3:")
print(classification_report(y, y_pred_30, target_names=['Neg', 'Pos']))
```

## Key Observations
- class_weight='balanced' upweights minority
- Lower threshold: higher recall, lower precision
- Choose threshold based on cost of FP vs FN

# Local vs Global Minima

Explore how initialization affects which minimum gradient descent finds in non-convex functions.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Non-convex function with multiple minima
f = lambda x: x**4 - 4*x**2 + x
df = lambda x: 4*x**3 - 8*x + 1

x = np.linspace(-2.5, 2.5, 200)
plt.figure(figsize=(10, 5))
plt.plot(x, f(x), 'b-', linewidth=2)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Non-convex function with two local minima')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
def gradient_descent(df, x0, lr=0.1, n_steps=100):
    x = x0
    history = [x]
    for _ in range(n_steps):
        x = x - lr * df(x)
        history.append(x)
    return x, np.array(history)

# Different starting points
starts = [-2.0, -0.5, 0.5, 2.0]
results = {}
for s in starts:
    x_min, hist = gradient_descent(df, s, lr=0.01, n_steps=200)
    results[s] = (x_min, hist)
    print(f"Start: {s:5.1f} → Minimum: {x_min:.4f}, f(x) = {f(x_min):.4f}")
```

```python
plt.figure(figsize=(10, 5))
plt.plot(x, f(x), 'b-', linewidth=2, label='f(x)')

colors = ['red', 'green', 'orange', 'purple']
for (s, (xm, hist)), c in zip(results.items(), colors):
    plt.scatter(hist[::10], f(hist[::10]), c=c, s=30, alpha=0.7)
    plt.scatter(xm, f(xm), c=c, s=100, marker='*', label=f'Start {s}')

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Different starts → Different minima')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Random restarts to find global minimum
np.random.seed(42)
best_x, best_f = None, np.inf

for _ in range(20):
    x0 = np.random.uniform(-2.5, 2.5)
    xm, _ = gradient_descent(df, x0, lr=0.01, n_steps=200)
    if f(xm) < best_f:
        best_x, best_f = xm, f(xm)

print(f"\nBest found with random restarts:")
print(f"  x = {best_x:.4f}, f(x) = {best_f:.4f}")
```

## Key Observations
- Non-convex functions have multiple local minima
- Gradient descent finds nearest local minimum from starting point
- Different initializations lead to different minima
- Random restarts help find global minimum
- This is why neural network initialization matters

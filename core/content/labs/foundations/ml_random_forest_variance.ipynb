# Random Forest Variance Reduction

Show how Random Forest reduces variance compared to single tree.

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
np.random.seed(42)

X, y = make_classification(n_samples=500, n_features=10, random_state=42)
```

```python
# Variance across different random states
tree_scores = []
rf_scores = []

for seed in range(20):
    tree = DecisionTreeClassifier(random_state=seed)
    rf = RandomForestClassifier(n_estimators=50, random_state=seed)
    
    tree_scores.append(cross_val_score(tree, X, y, cv=3).mean())
    rf_scores.append(cross_val_score(rf, X, y, cv=3).mean())

print(f"Tree: mean={np.mean(tree_scores):.3f}, std={np.std(tree_scores):.3f}")
print(f"RF:   mean={np.mean(rf_scores):.3f}, std={np.std(rf_scores):.3f}")
```

## Key Observations
- Random Forest averages many decorrelated trees
- Lower variance than single tree
- Feature and sample randomization reduce correlation
- More trees generally = more stable

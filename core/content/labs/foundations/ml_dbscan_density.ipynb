# DBSCAN Density Clustering

Compare K-Means and DBSCAN on non-spherical clusters.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import KMeans, DBSCAN
np.random.seed(42)

X, _ = make_moons(n_samples=300, noise=0.05, random_state=42)
```

```python
km = KMeans(n_clusters=2, random_state=42).fit(X)
db = DBSCAN(eps=0.2, min_samples=5).fit(X)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].scatter(X[:, 0], X[:, 1], c=km.labels_, cmap='viridis')
axes[0].set_title('K-Means (fails on moons)')
axes[1].scatter(X[:, 0], X[:, 1], c=db.labels_, cmap='viridis')
axes[1].set_title('DBSCAN (handles moons)')
plt.show()
```

## Key Observations
- K-Means assumes spherical clusters
- DBSCAN: density-based, finds arbitrary shapes
- DBSCAN params: eps (neighborhood radius), min_samples
- Label -1 = noise points

# Duplicate and Outlier Handling

Detect and handle duplicates and outliers in data.

```python
import numpy as np
import pandas as pd
np.random.seed(42)

df = pd.DataFrame({'A': np.random.randn(100), 'B': np.random.randn(100)})
df = pd.concat([df, df.iloc[:5]])  # Add duplicates
df.iloc[0, 0] = 100  # Add outlier
print(f"Shape: {df.shape}")
```

```python
# Detect duplicates
print(f"Duplicates: {df.duplicated().sum()}")
df_dedup = df.drop_duplicates()
print(f"After dedup: {df_dedup.shape}")

# Detect outliers (IQR)
Q1, Q3 = df_dedup['A'].quantile([0.25, 0.75])
IQR = Q3 - Q1
outliers = df_dedup[(df_dedup['A'] < Q1 - 1.5*IQR) | (df_dedup['A'] > Q3 + 1.5*IQR)]
print(f"\nOutliers in A: {len(outliers)}")

# Cap outliers
df_capped = df_dedup.copy()
df_capped['A'] = df_capped['A'].clip(Q1 - 1.5*IQR, Q3 + 1.5*IQR)
print(f"A range after capping: [{df_capped['A'].min():.2f}, {df_capped['A'].max():.2f}]")
```

## Key Observations
- drop_duplicates() removes exact duplicates
- IQR/Z-score for outlier detection
- Options: remove, cap, or transform outliers

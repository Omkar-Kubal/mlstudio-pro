# Missing Data Handling

Detect, analyze, and handle missing values in DataFrames.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': [1, 2, np.nan, 4],
    'B': [5, np.nan, np.nan, 8],
    'C': ['a', 'b', 'c', None]
})
print(df)
```

```python
# Detect missing
print(f"Is null:\n{df.isnull()}")
print(f"\nCounts: {df.isnull().sum().to_dict()}")
print(f"Percentage: {(df.isnull().sum() / len(df) * 100).round(1).to_dict()}")
```

```python
# Drop missing
print(f"Drop rows with any NaN:\n{df.dropna()}")
print(f"\nDrop rows where all NaN:\n{df.dropna(how='all')}")
print(f"\nDrop cols with NaN:\n{df.dropna(axis=1)}")
```

```python
# Fill missing
print(f"Fill with 0:\n{df.fillna(0)}")
print(f"\nFill A with mean:\n{df['A'].fillna(df['A'].mean())}")
print(f"\nForward fill:\n{df.fillna(method='ffill')}")
```

```python
# Interpolation
ts = pd.Series([1, np.nan, np.nan, 4, 5])
print(f"Original: {ts.tolist()}")
print(f"Interpolated: {ts.interpolate().tolist()}")
```

## Key Observations
- isnull()/isna() detect missing values
- dropna() removes, fillna() replaces
- Common fills: mean, median, mode, or forward/backward
- interpolate() for time series or ordered data
